Shipped daily-compare checkpoint: **h=96, L=2, heads=2** (`da_gps_chunked_l4_mvagg_gine_metaaux_regce_20260516_225149_CCE`; the `l4` folder name is legacy).

````python
# --- Lighter DA-GPS training (long-running; run on Colab GPU or local CUDA) ---
import os
import sys
import subprocess
import datetime
import warnings
from pathlib import Path

# Colab: mount Drive when chunks/caches live under MyDrive (uncomment on first run):
# from google.colab import drive
# drive.mount("/content/drive")

DRIVE_ROOT = Path("/content/drive")
MYDRIVE_DATA = DRIVE_ROOT / "MyDrive/datasets_gnn2"
COLAB_CHUNK_DEFAULT = MYDRIVE_DATA / "original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
WIN_CHUNK_DEFAULT = Path(r"D:\datasets\original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")


def _is_windows_drive_path(path) -> bool:
    s = str(path).strip().replace("/", "\\")
    return len(s) >= 2 and s[1] == ":" and s[0].isalpha()


def _on_colab() -> bool:
    return Path("/content").is_dir()


def _drive_mounted() -> bool:
    return DRIVE_ROOT.is_dir()


def _resolve_data_path(path: Path, *, label: str, colab_fallback: Path | None = None) -> Path:
    """Resolve a data path; never join Windows drive letters with cwd on Linux."""
    raw = str(path)
    if _is_windows_drive_path(raw):
        if _on_colab() and colab_fallback is not None:
            warnings.warn(
                f"{label}={raw!r} is a Windows path on Linux/Colab; using {colab_fallback} instead.",
                UserWarning,
                stacklevel=2,
            )
            return colab_fallback.expanduser().resolve()
        if _on_colab():
            raise ValueError(
                f"{label}={raw!r} is a Windows absolute path and invalid on Colab. "
                f"Mount Drive and use e.g. {COLAB_CHUNK_DEFAULT}"
            )
        return Path(raw).expanduser().resolve()
    p = Path(path).expanduser()
    if p.is_absolute():
        return p.resolve()
    return (Path.cwd() / p).resolve()


def _find_node_pe_csv(chunk_parent: Path) -> Path:
    hits = sorted(chunk_parent.glob("run_*/gnn_node_index_master.csv"))
    if not hits:
        raise FileNotFoundError(
            f"No run_*/gnn_node_index_master.csv under {chunk_parent}"
        )
    return hits[0]


# --- paths: auto-detect Colab + Drive; override below if needed ---
_env_repo = os.environ.get("GNN2_REPO_ROOT", "").strip()
REPO = Path(_env_repo).expanduser().resolve() if _env_repo else Path.cwd().resolve()
if not (REPO / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
    raise FileNotFoundError(f"Repo root missing trainer script: {REPO}")

if _drive_mounted():
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    DA_CACHE_ROOT = MYDRIVE_DATA / "cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = MYDRIVE_DATA / "runs"
elif _on_colab():
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    DA_CACHE_ROOT = MYDRIVE_DATA / "cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
elif os.name == "nt":
    CHUNK_PARENT = WIN_CHUNK_DEFAULT
    DA_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
else:
    CHUNK_PARENT = REPO / "datasets_gnn2_from pc/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
    DA_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"

NODE_PE_CSV = None  # auto: first run_*/gnn_node_index_master.csv under CHUNK_PARENT

# Optional overrides (use POSIX paths on Colab, not D:\...):
# CHUNK_PARENT = Path("/content/drive/MyDrive/datasets_gnn2/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")
# NODE_PE_CSV = CHUNK_PARENT / "run_001_scen_0000_0049_seed_20420233/gnn_node_index_master.csv"

# --- lighter architecture (sweeps: LIGHTER_LAYERS=1, LIGHTER_HIDDEN=48) ---
LIGHTER_HIDDEN = 64
LIGHTER_LAYERS = 2
LIGHTER_HEADS = 2
LIGHTER_NODE_EMB_DIM = 2  # current shipped model used 4 at h=96

META_AUX_COLS = "pv_pv2_p_post_kw,pv_pv2_q_post_kvar,P_loss_total_post_kw,Q_loss_total_post_kvar"
NUM_WORKERS = 0 if os.name == "nt" else 4

os.chdir(REPO)
os.environ["PYTHONUNBUFFERED"] = "1"
os.environ.setdefault("GNN2_REPO_ROOT", str(REPO))

_colab_fb = COLAB_CHUNK_DEFAULT if _on_colab() else None
chunk_parent = _resolve_data_path(CHUNK_PARENT, label="CHUNK_PARENT", colab_fallback=_colab_fb)
da_cache_root = _resolve_data_path(
    DA_CACHE_ROOT,
    label="DA_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / "cache/da_gps_chunked_mvagg_full_gine" if _on_colab() else None,
)
gnn_cache_root = _resolve_data_path(
    GNN_CACHE_ROOT,
    label="GNN_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine" if _on_colab() else None,
)
runs_parent = _resolve_data_path(
    RUNS_PARENT,
    label="RUNS_PARENT",
    colab_fallback=MYDRIVE_DATA / "runs" if _on_colab() else None,
)

if _on_colab() and not _drive_mounted():
    print("Colab detected but /content/drive is not mounted — uncomment drive.mount() above.")

print("=== Preflight ===")
print(f"REPO:           {REPO}")
print(f"CHUNK_PARENT:   {chunk_parent}")
if not chunk_parent.is_dir():
    raise FileNotFoundError(f"CHUNK_PARENT not found: {chunk_parent}")

run_preview = sorted(
    p.name for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_")
)[:5]
print(f"run_* preview:  {run_preview}")

if NODE_PE_CSV is None:
    node_pe = _find_node_pe_csv(chunk_parent)
else:
    node_pe = _resolve_data_path(NODE_PE_CSV, label="NODE_PE_CSV", colab_fallback=None)
    if not node_pe.is_file():
        warnings.warn(
            f"NODE_PE_CSV not found at {node_pe}; auto-discovering under CHUNK_PARENT.",
            UserWarning,
            stacklevel=2,
        )
        node_pe = _find_node_pe_csv(chunk_parent)

print(f"NODE_PE_CSV:    {node_pe}")
print(f"DA_CACHE_ROOT:  {da_cache_root}")
print(f"GNN_CACHE_ROOT: {gnn_cache_root}")
print(f"RUNS_PARENT:    {runs_parent}")
print("=================\n")

da_cache_root.mkdir(parents=True, exist_ok=True)
gnn_cache_root.mkdir(parents=True, exist_ok=True)
runs_parent.mkdir(parents=True, exist_ok=True)

chunks = sorted(p for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_"))
if not chunks:
    raise RuntimeError(f"No run_* folders under {chunk_parent}")

for p in chunks:
    for name in (
        "gnn_node_features_and_targets_mvagg.csv",
        "gnn_edges_phase_static.csv",
        "gnn_sample_meta.csv",
    ):
        if not (p / name).is_file():
            raise FileNotFoundError(f"Missing {name} in {p.name}")

print(f"Found {len(chunks)} chunks under {chunk_parent}")

tag = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
out_dir = runs_parent / (
    f"da_gps_chunked_l{LIGHTER_LAYERS}_h{LIGHTER_HIDDEN}_mvagg_gine_metaaux_regce_{tag}"
)
out_dir.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable, "-u", "train_da_gps_multitask_complex_voltage_gine.py",
    "--chunk_parent", str(chunk_parent),
    "--chunk_subdir_glob", "run_*",
    "--nodes_csv", "gnn_node_features_and_targets_mvagg.csv",
    "--edge_catalog_csv", "gnn_edges_phase_static.csv",
    "--meta_csv", "gnn_sample_meta.csv",
    "--node_feature_cols", "p_load_kw,q_load_kvar,p_pv_kw",
    "--exclude_bess_features",
    "--node_pe_csv", str(node_pe),
    "--node_pe_cols", "auto",
    "--n_system_tokens", "10",
    "--aux_meta_cols", META_AUX_COLS,
    "--lambda_pv", "0.1",
    "--out_dir", str(out_dir),
    "--cache_dir", str(da_cache_root),
    "--bootstrap_gnn_cache_dir", str(gnn_cache_root),
    "--epochs", "200",
    "--batch_size", "64",
    "--hidden", str(LIGHTER_HIDDEN),
    "--layers", str(LIGHTER_LAYERS),
    "--heads", str(LIGHTER_HEADS),
    "--node_emb_dim", str(LIGHTER_NODE_EMB_DIM),
    "--edge_emb_dim", "0",
    "--lr", "5e-4",
    "--weight_decay", "1e-5",
    "--lambda_cap", "0.1",
    "--lambda_reg", "0.1",
    "--reg_loss", "ce",
    "--per_device_cap_head",
    "--per_device_reg_head",
    "--patience", "30",
    "--seed", "42",
    "--train_frac", "0.80",
    "--val_frac", "0.10",
    "--sample_frac", "1.0",
    "--num_workers", str(NUM_WORKERS),
    "--log_every", "10",
    "--checkpoint_every", "10",
    "--early_stop_on", "total",
    "--dropout", "0.1",
]

print("\nRunning:\n ", " ".join(cmd), "\n", flush=True)

with subprocess.Popen(
    cmd,
    cwd=str(REPO),
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    bufsize=1,
    text=True,
) as proc:
    assert proc.stdout is not None
    for line in proc.stdout:
        print(line, end="")
    rc = proc.wait()

if rc != 0:
    raise subprocess.CalledProcessError(rc, cmd)

print("\nTraining completed.")
print("Run dir:", out_dir.resolve())
print("Checkpoint (best):", (out_dir / "da_gps_multitask_best.pt").resolve())
print("Checkpoint (last):", (out_dir / "training_last.pt").resolve())
print("Regulator classes:", (out_dir / "reg_class_tables.json").resolve())
print("Report:", (out_dir / "da_gps_report.json").resolve())
````

## 4. DA-GPS training with physics-informed loss (optional)

Adds an optional **nodal P/Q power-balance** penalty on denormalized complex voltage predictions. After the voltage head, the trainer builds `Y_bus` from **line-only** `R_full`/`X_full` (regulator transformer branches in the edge CSV are skipped to avoid double-stamping), stamps **predicted** regulator tap ratios on branches from `hetero_mv_edge_catalog.csv`, and adds capacitor shunt `j·B` at mapped cap buses from **predicted** cap states (caps are in `Y` only, not in `Q_inj`). Nodal injections use **known** denormalized node features (OpenDSS convention: `P_inj = P_pv − P_load`, `Q_inj = −Q_pv − Q_load`); meta-aux PV totals are supervised separately and are **not** fed into the physics residual. Residuals `r_P = P_inj − Re(V·conj(YV))`, `r_Q = Q_inj − Im(V·conj(YV))` are averaged on MV nodes (`--pf_balance_nodes mv`: `electrical_distance_ohm > 0`, slack/source buses excluded).

**Strict — no silent fallbacks.** Training fails at startup if any required artifact is missing or misconfigured.

### Requirements

| Item | Location / flag |
|------|-----------------|
| Regulator branch catalog | `hetero_mv_edge_catalog.csv` → `--pf_reg_edge_catalog` (or under first chunk / mvagg root) |
| Capacitor bus map | `capacitor_involved_nodes.csv` → `--pf_cap_nodes_csv` (shunt in `Y` only) |
| Meta-aux PV columns | Same as section 3 for **supervised** `loss_pv` only (not used in physics residual) |
| Node features | `p_load_kw`, `q_load_kvar`, `p_pv_kw` (and optionally `q_pv_kvar`) in `--node_feature_cols` |
| MV balance mask | `electrical_distance_ohm` column in `gnn_node_features_and_targets_mvagg.csv` |

Default weight: `--loss_power_balance_weight 0.01` (try `0.001` if loss dominates or training destabilizes). Optional `--pf_detach_controls` stops gradients through predicted taps/caps inside the physics term.

````python
# --- DA-GPS training with physics-informed P/Q balance loss (long-running) ---
import os
import sys
import subprocess
import datetime
import warnings
from pathlib import Path

# Colab: mount Drive when chunks/caches live under MyDrive (uncomment on first run):
# from google.colab import drive
# drive.mount("/content/drive")

DRIVE_ROOT = Path("/content/drive")
MYDRIVE_DATA = DRIVE_ROOT / "MyDrive/datasets_gnn2"
COLAB_CHUNK_DEFAULT = MYDRIVE_DATA / "original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
WIN_CHUNK_DEFAULT = Path(r"D:\datasets\original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")

# mvagg static topology for PF loss (regulator edges + cap bus map)
COLAB_PF_DATA_DEFAULT = MYDRIVE_DATA / "loadtype_8500_dailyagg"
WIN_PF_DATA_DEFAULT = Path(r"D:\datasets\loadtype_8500_dailyagg")


def _is_windows_drive_path(path) -> bool:
    s = str(path).strip().replace("/", "\\")
    return len(s) >= 2 and s[1] == ":" and s[0].isalpha()


def _on_colab() -> bool:
    return Path("/content").is_dir()


def _drive_mounted() -> bool:
    return DRIVE_ROOT.is_dir()


def _resolve_data_path(path: Path, *, label: str, colab_fallback: Path | None = None) -> Path:
    raw = str(path)
    if _is_windows_drive_path(raw):
        if _on_colab() and colab_fallback is not None:
            warnings.warn(
                f"{label}={raw!r} is a Windows path on Linux/Colab; using {colab_fallback} instead.",
                UserWarning,
                stacklevel=2,
            )
            return colab_fallback.expanduser().resolve()
        if _on_colab():
            raise ValueError(
                f"{label}={raw!r} is a Windows absolute path and invalid on Colab. "
                f"Mount Drive and use e.g. {COLAB_CHUNK_DEFAULT}"
            )
        return Path(raw).expanduser().resolve()
    p = Path(path).expanduser()
    if p.is_absolute():
        return p.resolve()
    return (Path.cwd() / p).resolve()


def _find_node_pe_csv(chunk_parent: Path) -> Path:
    hits = sorted(chunk_parent.glob("run_*/gnn_node_index_master.csv"))
    if not hits:
        raise FileNotFoundError(
            f"No run_*/gnn_node_index_master.csv under {chunk_parent}"
        )
    return hits[0]


def _resolve_pf_catalog_paths(pf_data_root: Path) -> tuple[Path, Path]:
    reg = pf_data_root / "Heterogenous GNN dataset" / "edges" / "hetero_mv_edge_catalog.csv"
    cap = pf_data_root / "capacitor_involved_nodes.csv"
    missing = [p for p in (reg, cap) if not p.is_file()]
    if missing:
        raise FileNotFoundError(
            "Physics loss requires mvagg topology CSVs:\n  "
            + "\n  ".join(str(p) for p in missing)
            + f"\nSet PF_DATA_ROOT to the folder containing loadtype_8500_dailyagg artifacts."
        )
    return reg.resolve(), cap.resolve()


# --- paths: auto-detect Colab + Drive; override below if needed ---
_env_repo = os.environ.get("GNN2_REPO_ROOT", "").strip()
REPO = Path(_env_repo).expanduser().resolve() if _env_repo else Path.cwd().resolve()
if not (REPO / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
    raise FileNotFoundError(f"Repo root missing trainer script: {REPO}")

if _drive_mounted():
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    PF_DATA_ROOT = COLAB_PF_DATA_DEFAULT
    DA_CACHE_ROOT = MYDRIVE_DATA / "cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = MYDRIVE_DATA / "runs"
elif _on_colab():
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    PF_DATA_ROOT = COLAB_PF_DATA_DEFAULT
    DA_CACHE_ROOT = MYDRIVE_DATA / "cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
elif os.name == "nt":
    CHUNK_PARENT = WIN_CHUNK_DEFAULT
    PF_DATA_ROOT = WIN_PF_DATA_DEFAULT
    DA_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
else:
    CHUNK_PARENT = REPO / "datasets_gnn2_from pc/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
    PF_DATA_ROOT = REPO / "datasets_gnn2_from pc/loadtype_8500_dailyagg"
    DA_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"

NODE_PE_CSV = None  # auto: first run_*/gnn_node_index_master.csv under CHUNK_PARENT

# --- architecture (match section 3 defaults; sweeps: LIGHTER_LAYERS=1, LIGHTER_HIDDEN=48) ---
LIGHTER_HIDDEN = 64
LIGHTER_LAYERS = 2
LIGHTER_HEADS = 2
LIGHTER_NODE_EMB_DIM = 2

META_AUX_COLS = "pv_pv2_p_post_kw,pv_pv2_q_post_kvar,P_loss_total_post_kw,Q_loss_total_post_kvar"
NUM_WORKERS = 0 if os.name == "nt" else 4

# --- physics loss knobs ---
RUN_PF_VALIDATE = True  # runs _tmp_pf_physics_validate.py before training
LOSS_POWER_BALANCE_WEIGHT = 0.01  # tune down to 0.001 if physics term dominates
PF_BALANCE_NODES = "mv"  # mv = electrical_distance_ohm > 0; or "all"
PF_S_BASE_KVA = 5000.0
PF_KV_BASE = 12.47
PF_DETACH_CONTROLS = False  # set True for stability (no grad through taps/caps in physics term)

os.chdir(REPO)
os.environ["PYTHONUNBUFFERED"] = "1"
os.environ.setdefault("GNN2_REPO_ROOT", str(REPO))

_colab_fb = COLAB_CHUNK_DEFAULT if _on_colab() else None
chunk_parent = _resolve_data_path(CHUNK_PARENT, label="CHUNK_PARENT", colab_fallback=_colab_fb)
pf_data_root = _resolve_data_path(
    PF_DATA_ROOT,
    label="PF_DATA_ROOT",
    colab_fallback=COLAB_PF_DATA_DEFAULT if _on_colab() else None,
)
da_cache_root = _resolve_data_path(
    DA_CACHE_ROOT,
    label="DA_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / "cache/da_gps_chunked_mvagg_full_gine" if _on_colab() else None,
)
gnn_cache_root = _resolve_data_path(
    GNN_CACHE_ROOT,
    label="GNN_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine" if _on_colab() else None,
)
runs_parent = _resolve_data_path(
    RUNS_PARENT,
    label="RUNS_PARENT",
    colab_fallback=MYDRIVE_DATA / "runs" if _on_colab() else None,
)

pf_reg_catalog, pf_cap_nodes_csv = _resolve_pf_catalog_paths(pf_data_root)

if _on_colab() and not _drive_mounted():
    print("Colab detected but /content/drive is not mounted — uncomment drive.mount() above.")

print("=== Preflight (physics-informed) ===")
print(f"REPO:              {REPO}")
print(f"CHUNK_PARENT:      {chunk_parent}")
print(f"PF_DATA_ROOT:      {pf_data_root}")
print(f"PF reg catalog:    {pf_reg_catalog}")
print(f"PF cap nodes CSV:  {pf_cap_nodes_csv}")
if not chunk_parent.is_dir():
    raise FileNotFoundError(f"CHUNK_PARENT not found: {chunk_parent}")

run_preview = sorted(
    p.name for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_")
)[:5]
print(f"run_* preview:     {run_preview}")

if NODE_PE_CSV is None:
    node_pe = _find_node_pe_csv(chunk_parent)
else:
    node_pe = _resolve_data_path(NODE_PE_CSV, label="NODE_PE_CSV", colab_fallback=None)
    if not node_pe.is_file():
        warnings.warn(
            f"NODE_PE_CSV not found at {node_pe}; auto-discovering under CHUNK_PARENT.",
            UserWarning,
            stacklevel=2,
        )
        node_pe = _find_node_pe_csv(chunk_parent)

print(f"NODE_PE_CSV:       {node_pe}")
print(f"DA_CACHE_ROOT:     {da_cache_root}")
print(f"GNN_CACHE_ROOT:    {gnn_cache_root}")
print(f"RUNS_PARENT:       {runs_parent}")
print(f"loss_power_balance: {LOSS_POWER_BALANCE_WEIGHT}")
print("==================================\n")

if RUN_PF_VALIDATE:
    validate_script = REPO / "_tmp_pf_physics_validate.py"
    if not validate_script.is_file():
        raise FileNotFoundError(
            f"Missing {validate_script.name} — git pull commit c932f84+ or set RUN_PF_VALIDATE=False"
        )
    print("Running PF physics validation (_tmp_pf_physics_validate.py)...\n", flush=True)
    subprocess.run([sys.executable, "-u", str(validate_script)], cwd=str(REPO), check=True)
    print("\nPF validation passed.\n", flush=True)

da_cache_root.mkdir(parents=True, exist_ok=True)
gnn_cache_root.mkdir(parents=True, exist_ok=True)
runs_parent.mkdir(parents=True, exist_ok=True)

chunks = sorted(p for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_"))
if not chunks:
    raise RuntimeError(f"No run_* folders under {chunk_parent}")

for p in chunks:
    for name in (
        "gnn_node_features_and_targets_mvagg.csv",
        "gnn_edges_phase_static.csv",
        "gnn_sample_meta.csv",
    ):
        if not (p / name).is_file():
            raise FileNotFoundError(f"Missing {name} in {p.name}")

print(f"Found {len(chunks)} chunks under {chunk_parent}")

tag = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
out_dir = runs_parent / (
    f"da_gps_chunked_l{LIGHTER_LAYERS}_h{LIGHTER_HIDDEN}_mvagg_gine_metaaux_regce_pfphys_{tag}"
)
out_dir.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable, "-u", "train_da_gps_multitask_complex_voltage_gine.py",
    "--chunk_parent", str(chunk_parent),
    "--chunk_subdir_glob", "run_*",
    "--nodes_csv", "gnn_node_features_and_targets_mvagg.csv",
    "--edge_catalog_csv", "gnn_edges_phase_static.csv",
    "--meta_csv", "gnn_sample_meta.csv",
    "--node_feature_cols", "p_load_kw,q_load_kvar,p_pv_kw",
    "--exclude_bess_features",
    "--node_pe_csv", str(node_pe),
    "--node_pe_cols", "auto",
    "--n_system_tokens", "10",
    "--aux_meta_cols", META_AUX_COLS,
    "--lambda_pv", "0.1",
    "--out_dir", str(out_dir),
    "--cache_dir", str(da_cache_root),
    "--bootstrap_gnn_cache_dir", str(gnn_cache_root),
    "--epochs", "200",
    "--batch_size", "64",
    "--hidden", str(LIGHTER_HIDDEN),
    "--layers", str(LIGHTER_LAYERS),
    "--heads", str(LIGHTER_HEADS),
    "--node_emb_dim", str(LIGHTER_NODE_EMB_DIM),
    "--edge_emb_dim", "0",
    "--lr", "5e-4",
    "--weight_decay", "1e-5",
    "--lambda_cap", "0.1",
    "--lambda_reg", "0.1",
    "--reg_loss", "ce",
    "--per_device_cap_head",
    "--per_device_reg_head",
    "--patience", "30",
    "--seed", "42",
    "--train_frac", "0.80",
    "--val_frac", "0.10",
    "--sample_frac", "1.0",
    "--num_workers", str(NUM_WORKERS),
    "--log_every", "10",
    "--checkpoint_every", "10",
    "--early_stop_on", "total",
    "--dropout", "0.1",
    # --- physics-informed loss (strict; commit c932f84+) ---
    "--loss_power_balance_weight", str(LOSS_POWER_BALANCE_WEIGHT),
    "--pf_balance_nodes", PF_BALANCE_NODES,
    "--pf_s_base_kva", str(PF_S_BASE_KVA),
    "--pf_kv_base", str(PF_KV_BASE),
    "--pf_reg_edge_catalog", str(pf_reg_catalog),
    "--pf_cap_nodes_csv", str(pf_cap_nodes_csv),
]
if PF_DETACH_CONTROLS:
    cmd.append("--pf_detach_controls")

print("\nRunning:\n ", " ".join(cmd), "\n", flush=True)

with subprocess.Popen(
    cmd,
    cwd=str(REPO),
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    bufsize=1,
    text=True,
) as proc:
    assert proc.stdout is not None
    for line in proc.stdout:
        print(line, end="")
    rc = proc.wait()

if rc != 0:
    raise subprocess.CalledProcessError(rc, cmd)

print("\nTraining completed.")
print("Run dir:", out_dir.resolve())
print("Checkpoint (best):", (out_dir / "da_gps_multitask_best.pt").resolve())
print("Checkpoint (last):", (out_dir / "training_last.pt").resolve())
print("Regulator classes:", (out_dir / "reg_class_tables.json").resolve())
print("Report:", (out_dir / "da_gps_report.json").resolve())
````
