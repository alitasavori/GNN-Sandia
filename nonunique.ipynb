## 3. Train lighter DA-GPS variants (optional — long-running)

Shipped daily-compare checkpoint: **h=96, L=2, heads=2** (`da_gps_chunked_l4_mvagg_gine_metaaux_regce_20260516_225149_CCE`; the `l4` folder name is legacy).

The execution cell below trains **smaller** GINE stacks (default **h=64, L=2, heads=2, node_emb_dim=2**) with the same recipe as that run: mvagg nodes, regulator CE (`--reg_loss ce`), meta_aux, `--exclude_bess_features`, chunked `run_*` folders. Goal: trade a bit of voltage MAE for faster per-step inference — re-run **cell 2** afterward to compare timing and MAE.

**Not for `run_and_plot`:** this calls `train_da_gps_multitask_complex_voltage_gine.py` directly (hours on GPU; tensor-cache build on first run).

### Paths (Colab vs local)

| | Colab (Drive mounted) | Local Windows |
|---|------------------------|---------------|
| Repo | `/content/GNN-Sandia` or `/content/GNN2` | notebook cwd or `GNN2_REPO_ROOT` |
| `CHUNK_PARENT` | `/content/drive/MyDrive/datasets_gnn2/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40` | `D:\datasets\original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40` |
| `NODE_PE_CSV` | auto: first `run_*/gnn_node_index_master.csv` under `CHUNK_PARENT` | same |
| Caches / runs | under `.../datasets_gnn2/cache/` and `.../runs/` on Drive | repo `datasets_gnn2_from pc/cache/` and `gnn2_architecture_search/attention checkpoints/` |

On Colab: the training cell **auto-mounts Drive** and **ignores Windows `D:\` paths** (they would wrongly resolve under the repo on Linux). A preflight block prints resolved paths and a preview of `run_*` folders before training starts.

### Checkpoint persistence (Colab — read before full training)

**Warning:** Checkpoints under `/content/GNN-Sandia/gnn2_architecture_search/attention checkpoints/` are **ephemeral**. A runtime disconnect or factory reset deletes them.

Section 3 defaults `--out_dir` to **Google Drive** when Colab + Drive are detected:

`/content/drive/MyDrive/datasets_gnn2/runs/da_gps_chunked_l{L}_h{H}_mvagg_gine_metaaux_regce_{timestamp}/`

(full run: `SMOKE_TEST=False`, 40 chunks, 200 epochs → no `_smoke` suffix)

**Section 8** physics fine-tune requires that baseline run folder on Drive. Mount Drive before training; do not override `RUNS_PARENT` to the git clone on Colab.

### Physics-informed nodal loss (training toggle)

Set `PHYSICS_WEIGHT` at the top of the execution cell:

| `PHYSICS_WEIGHT` | Mode | CLI flags | Run folder suffix |
|---|---|---|---|
| `0.0` (default) | **Baseline** — no physics loss | omit all `--pf_*` / `--loss_power_balance_weight` | `..._metaaux_regce_{tag}` |
| `0.01` | **Physics-informed** — sparse Y-bus Huber residual on predicted voltages during **training only** (controls detached) | `--loss_power_balance_weight 0.01 --pf_sparse_y 1 --pf_huber_delta_kw 10 --pf_detach_controls` | `..._metaaux_regce_pf_{tag}` |

**Section 6** below is separate **offline verification** (surrogate Y@V vs dataset P_inj/Q_inj at label V on a snapshot; no OpenDSS).

### PF topology CSVs (required when PHYSICS_WEIGHT > 0)

After git pull, the training cell resolves PF files from repo **colab_pf_data/** (~1.3 MB bundle) via

### Explicit PF balance nodes (optional)

Set `PF_BALANCE_NODE_LIST` in the training cell to a CSV path (e.g. repo `colab_pf_data/pf_balance_nodes_explicit.csv`, 1177 hetero MV load nodes (node-name keyed; chunk-safe)) to **skip** automatic MV mask refinement (distance + hetero + interface + Y-closure). Leave `None` for legacy `--pf_balance_nodes mv` behavior.

| `PF_BALANCE_NODE_LIST` | Behavior |
|---|---|
| `None` | Legacy: `--pf_balance_nodes mv` + hetero/interface refinement |
| path to CSV | `--pf_balance_node_list_csv PATH` only (columns: `node_idx`, `bus`, and/or `node`) |


### Smoke test vs full training

| `SMOKE_TEST` | Chunks | `epochs` | `patience` | Cache |
|---|---|---|---|---|
| `True` (default) | first `SMOKE_CHUNK_COUNT` `run_*` (default **3**) | `SMOKE_EPOCHS` (**15**) | `SMOKE_PATIENCE` (**5**) | `da_gps_chunked_mvagg_smoke_gine` |
| `False` | all `run_*` (40) | `200` | `30` | `da_gps_chunked_mvagg_full_gine` |

**Tradeoff:** 3–5 chunks is more representative than 1-chunk smoke but much faster than 40 chunks. `SMOKE_EPOCHS=15` is a speed default — raise if metrics are noisy. Cells 6 (baseline) and 17 (physics) share the same chunk set, seed, and smoke cache; run **section 7** (compare cell) afterward to compare test |V| MAE.

Physics smoke cell (cell 17) defaults to `PHYSICS_WEIGHT = 0.05` (pu-scaled Huber; full runs may use `0.01`). Iterate weight / epochs until section 7 shows physics beating baseline on test |V| MAE.
When `PHYSICS_WEIGHT > 0`, the cell passes `--pf_balance_node_list_csv` pointing at repo `colab_pf_data/pf_balance_nodes_explicit.csv` (1177 hetero MV load balance nodes; skips auto distance/hetero mask refinement).
When PHYSICS_WEIGHT > 0, the cell uses `--early_stop_on voltage` so the physics term does not drive patience.


````python
# --- Lighter DA-GPS training (long-running; run on Colab GPU or local CUDA) ---
import os
import sys
import subprocess
import datetime
import warnings
from pathlib import Path

DRIVE_ROOT = Path("/content/drive")
MYDRIVE_DATA = DRIVE_ROOT / "MyDrive/datasets_gnn2"
COLAB_CHUNK_DEFAULT = MYDRIVE_DATA / "original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
WIN_CHUNK_DEFAULT = Path(r"D:\datasets\original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")


def _is_windows_drive_path(path) -> bool:
    s = str(path).strip().replace("/", "\\")
    return len(s) >= 2 and s[1] == ":" and s[0].isalpha()


def _on_colab() -> bool:
    return Path("/content").is_dir()


def _drive_mounted() -> bool:
    return DRIVE_ROOT.is_dir() and (DRIVE_ROOT / "MyDrive").is_dir()


def _resolve_data_path(path: Path, *, label: str, colab_fallback: Path | None = None) -> Path:
    """Resolve a data path; never join Windows drive letters with cwd on Linux."""
    raw = str(path)
    if _is_windows_drive_path(raw):
        if _on_colab() and colab_fallback is not None:
            warnings.warn(
                f"{label}={raw!r} is a Windows path on Linux/Colab; using {colab_fallback} instead.",
                UserWarning,
                stacklevel=2,
            )
            return colab_fallback.expanduser().resolve()
        if _on_colab():
            raise ValueError(
                f"{label}={raw!r} is a Windows absolute path and invalid on Colab. "
                f"Mount Drive and use e.g. {COLAB_CHUNK_DEFAULT}"
            )
        return Path(raw).expanduser().resolve()
    p = Path(path).expanduser()
    if p.is_absolute():
        return p.resolve()
    return (Path.cwd() / p).resolve()


def _find_node_pe_csv(chunk_parent: Path) -> Path:
    hits = sorted(chunk_parent.glob("run_*/gnn_node_index_master.csv"))
    if not hits:
        raise FileNotFoundError(
            f"No run_*/gnn_node_index_master.csv under {chunk_parent}"
        )
    return hits[0]

def _sorted_run_chunk_dirs(chunk_parent: Path) -> list[Path]:
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_")),
        key=lambda p: p.name,
    )


def _smoke_chunk_subdir_glob(chunk_parent: Path, count: int) -> str:
    """First `count` run_* names, comma-separated for --chunk_subdir_glob."""
    names = [p.name for p in _sorted_run_chunk_dirs(chunk_parent)]
    if len(names) < count:
        raise ValueError(
            f"SMOKE_CHUNK_COUNT={count} but only {len(names)} run_* under {chunk_parent}"
        )
    return ",".join(names[:count])


def _chunks_from_subdir_glob(chunk_parent: Path, glob_pat: str) -> list[Path]:
    import fnmatch

    glob_pat = str(glob_pat).strip()
    if "," in glob_pat:
        allowed = {s.strip() for s in glob_pat.split(",") if s.strip()}
        chunks = sorted(
            (p for p in chunk_parent.iterdir() if p.is_dir() and p.name in allowed),
            key=lambda p: p.name,
        )
        missing = allowed - {p.name for p in chunks}
        if missing:
            raise FileNotFoundError(f"Missing smoke chunk folders: {sorted(missing)}")
        return chunks
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and fnmatch.fnmatch(p.name, glob_pat)),
        key=lambda p: p.name,
    )


def _write_smoke_compare_manifest(
    runs_parent: Path,
    *,
    out_dir: Path,
    physics_weight: float,
    chunk_glob: str,
    n_chunks: int,
    epochs: int,
    seed: int,
) -> None:
    import json as _json

    path = runs_parent / "_smoke_compare_manifest.json"
    manifest = {}
    if path.is_file():
        manifest = _json.loads(path.read_text(encoding="utf-8"))
    key = "physics" if physics_weight > 0 else "baseline"
    manifest[key] = {
        "run_dir": str(out_dir.resolve()),
        "physics_weight": physics_weight,
        "chunk_glob": chunk_glob,
        "n_chunks": n_chunks,
        "epochs": epochs,
        "seed": seed,
    }
    path.write_text(_json.dumps(manifest, indent=2), encoding="utf-8")
    print(f"Smoke compare manifest ({key}) -> {path}")


# Colab: mount Drive when chunks/caches live under MyDrive
if _on_colab() and not _drive_mounted():
    from google.colab import drive
    drive.mount("/content/drive")

# --- smoke / full-run toggles (cells 6 baseline & 17 physics; cell 18 compares) ---
SMOKE_TEST = True
SMOKE_CHUNK_COUNT = 3  # first N run_* alphabetically; 3-5 = fast multi-chunk (not all 40)
SMOKE_EPOCHS = 15      # low for speed; more epochs = stabler MAE, ~linear time
SMOKE_PATIENCE = 5
SMOKE_SEED = 42
_DA_CACHE_NAME = "da_gps_chunked_mvagg_smoke_gine" if SMOKE_TEST else "da_gps_chunked_mvagg_full_gine"

# --- paths: auto-detect Colab + Drive; override below if needed ---
_env_repo = os.environ.get("GNN2_REPO_ROOT", "").strip()
REPO = Path(_env_repo).expanduser().resolve() if _env_repo else Path.cwd().resolve()
if not (REPO / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
    raise FileNotFoundError(f"Repo root missing trainer script: {REPO}")

if _on_colab():
    if not _drive_mounted():
        raise RuntimeError(
            "Colab requires Google Drive mounted. Re-run this cell and approve Drive access, "
            "or: from google.colab import drive; drive.mount('/content/drive')"
        )
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    DA_CACHE_ROOT = MYDRIVE_DATA / f"cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = MYDRIVE_DATA / "runs"  # persisted on Drive; /content/GNN-Sandia is ephemeral
elif os.name == "nt":
    CHUNK_PARENT = WIN_CHUNK_DEFAULT
    DA_CACHE_ROOT = REPO / f"datasets_gnn2_from pc/cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
else:
    CHUNK_PARENT = REPO / "datasets_gnn2_from pc/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
    DA_CACHE_ROOT = REPO / f"datasets_gnn2_from pc/cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"

NODE_PE_CSV = None  # auto: first run_*/gnn_node_index_master.csv under CHUNK_PARENT

# Optional overrides (use POSIX paths on Colab, not D:\...):
# CHUNK_PARENT = Path("/content/drive/MyDrive/datasets_gnn2/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")
# NODE_PE_CSV = CHUNK_PARENT / "run_001_scen_0000_0049_seed_20420233/gnn_node_index_master.csv"

# --- lighter architecture (sweeps: LIGHTER_LAYERS=1, LIGHTER_HIDDEN=48) ---
LIGHTER_HIDDEN = 64
LIGHTER_LAYERS = 2
LIGHTER_HEADS = 2
LIGHTER_NODE_EMB_DIM = 2  # current shipped model used 4 at h=96

PHYSICS_WEIGHT = 0.0  # baseline; use 0.01 for physics-informed


# Explicit PF balance nodes when physics on (1177 hetero MV load nodes (node-name keyed; chunk-safe); skips auto mask refinement)
PF_BALANCE_NODE_LIST_CSV = REPO / "colab_pf_data/pf_balance_nodes_explicit.csv"

META_AUX_COLS = "pv_pv2_p_post_kw,pv_pv2_q_post_kvar,P_loss_total_post_kw,Q_loss_total_post_kvar"
NUM_WORKERS = 0 if os.name == "nt" else 4

os.chdir(REPO)
sys.path.insert(0, str(REPO))
os.environ["PYTHONUNBUFFERED"] = "1"
os.environ.setdefault("GNN2_REPO_ROOT", str(REPO))

_colab_fb = COLAB_CHUNK_DEFAULT if _on_colab() else None
chunk_parent = _resolve_data_path(CHUNK_PARENT, label="CHUNK_PARENT", colab_fallback=_colab_fb)
if SMOKE_TEST:
    CHUNK_GLOB = _smoke_chunk_subdir_glob(chunk_parent, SMOKE_CHUNK_COUNT)
    EPOCHS = SMOKE_EPOCHS
    PATIENCE = SMOKE_PATIENCE
else:
    CHUNK_GLOB = "run_*"
    EPOCHS = 200
    PATIENCE = 30
da_cache_root = _resolve_data_path(
    DA_CACHE_ROOT,
    label="DA_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / f"cache/{_DA_CACHE_NAME}" if _on_colab() else None,
)
gnn_cache_root = _resolve_data_path(
    GNN_CACHE_ROOT,
    label="GNN_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine" if _on_colab() else None,
)
runs_parent = _resolve_data_path(
    RUNS_PARENT,
    label="RUNS_PARENT",
    colab_fallback=MYDRIVE_DATA / "runs" if _on_colab() else None,
)

# --- PF topology root (repo colab_pf_data/ after git pull, or Drive dailyagg fallback) ---
from gnn2_pf_data_paths import PF_CAP_NODES_REL, PF_REG_CATALOG_REL, resolve_pf_catalog_paths

PF_DATA_ROOT = None
if PHYSICS_WEIGHT > 0:
    _reg_cat, _cap_map, PF_DATA_ROOT = resolve_pf_catalog_paths(
        repo=REPO,
        preferred_root=None,
        chunk_parent=chunk_parent,
    )

print("=== Preflight ===")
print(f"REPO:           {REPO}")
print(f"SMOKE_TEST:     {SMOKE_TEST}")
print(f"SMOKE_CHUNK_COUNT: {SMOKE_CHUNK_COUNT}")
print(f"CHUNK_GLOB:     {CHUNK_GLOB}")
print(f"EPOCHS:         {EPOCHS}")
print(f"PATIENCE:       {PATIENCE}")
print(f"CHUNK_PARENT:   {chunk_parent}")
if not chunk_parent.is_dir():
    raise FileNotFoundError(f"CHUNK_PARENT not found: {chunk_parent}")

run_preview = sorted(
    p.name for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_")
)[:5]
print(f"run_* preview:  {run_preview}")

if NODE_PE_CSV is None:
    node_pe = _find_node_pe_csv(chunk_parent)
else:
    node_pe = _resolve_data_path(NODE_PE_CSV, label="NODE_PE_CSV", colab_fallback=None)
    if not node_pe.is_file():
        warnings.warn(
            f"NODE_PE_CSV not found at {node_pe}; auto-discovering under CHUNK_PARENT.",
            UserWarning,
            stacklevel=2,
        )
        node_pe = _find_node_pe_csv(chunk_parent)

print(f"NODE_PE_CSV:    {node_pe}")
print(f"DA_CACHE_ROOT:  {da_cache_root}")
print(f"GNN_CACHE_ROOT: {gnn_cache_root}")
print(f"RUNS_PARENT:    {runs_parent}")

if PHYSICS_WEIGHT > 0:
    assert PF_DATA_ROOT is not None
    _pf_checks = [
        ("reg_catalog", PF_DATA_ROOT / PF_REG_CATALOG_REL),
        ("cap_nodes", PF_DATA_ROOT / PF_CAP_NODES_REL),
        ("electrical_distance", PF_DATA_ROOT / "electrical_distance_from_substation.csv"),
        (
            "hetero_mv_nodes",
            PF_DATA_ROOT / "Heterogenous GNN dataset/nodes/hetero_mv_nodes_load_transformer.csv",
        ),
        ("bus_kv_cache", PF_DATA_ROOT / "bus_kv_base_by_node.csv"),
    ]
    print(f"PF_DATA_ROOT:   {PF_DATA_ROOT}")
    for label, p in _pf_checks:
        status = "OK" if p.is_file() else "MISSING"
        print(f"  PF {label}: {status}  {p}")
        if not p.is_file():
            raise FileNotFoundError(f"Physics preflight missing {label}: {p}")
else:
    print("PF_DATA_ROOT:   (skipped — PHYSICS_WEIGHT=0)")

print("=================")

da_cache_root.mkdir(parents=True, exist_ok=True)
gnn_cache_root.mkdir(parents=True, exist_ok=True)
runs_parent.mkdir(parents=True, exist_ok=True)

chunks = _chunks_from_subdir_glob(chunk_parent, CHUNK_GLOB)
if not chunks:
    raise RuntimeError(f"No folders for CHUNK_GLOB={CHUNK_GLOB!r} under {chunk_parent}")

for p in chunks:
    for name in (
        "gnn_node_features_and_targets_mvagg.csv",
        "gnn_edges_phase_static.csv",
        "gnn_sample_meta.csv",
    ):
        if not (p / name).is_file():
            raise FileNotFoundError(f"Missing {name} in {p.name}")

print(f"Found {len(chunks)} chunk(s) for CHUNK_GLOB under {chunk_parent}")

tag = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
_pf_suffix = "_pf" if PHYSICS_WEIGHT > 0 else ""
out_dir = runs_parent / (
    f"da_gps_chunked_l{LIGHTER_LAYERS}_h{LIGHTER_HIDDEN}_mvagg_gine_metaaux_regce{_pf_suffix}{'_smoke' if SMOKE_TEST else ''}_{tag}"
)
out_dir.mkdir(parents=True, exist_ok=True)

_early_stop = "voltage" if PHYSICS_WEIGHT > 0 else "total"

cmd = [
    sys.executable, "-u", "train_da_gps_multitask_complex_voltage_gine.py",
    "--chunk_parent", str(chunk_parent),
    "--chunk_subdir_glob", CHUNK_GLOB,
    "--nodes_csv", "gnn_node_features_and_targets_mvagg.csv",
    "--edge_catalog_csv", "gnn_edges_phase_static.csv",
    "--meta_csv", "gnn_sample_meta.csv",
    "--node_feature_cols", "p_load_kw,q_load_kvar,p_pv_kw",
    "--exclude_bess_features",
    "--node_pe_csv", str(node_pe),
    "--node_pe_cols", "auto",
    "--n_system_tokens", "10",
    "--aux_meta_cols", META_AUX_COLS,
    "--lambda_pv", "0.1",
    "--out_dir", str(out_dir),
    "--cache_dir", str(da_cache_root),
    "--bootstrap_gnn_cache_dir", str(gnn_cache_root),
    "--epochs", str(EPOCHS),
    "--batch_size", "64",
    "--hidden", str(LIGHTER_HIDDEN),
    "--layers", str(LIGHTER_LAYERS),
    "--heads", str(LIGHTER_HEADS),
    "--node_emb_dim", str(LIGHTER_NODE_EMB_DIM),
    "--edge_emb_dim", "0",
    "--lr", "5e-4",
    "--weight_decay", "1e-5",
    "--lambda_cap", "0.1",
    "--lambda_reg", "0.1",
    "--reg_loss", "ce",
    "--per_device_cap_head",
    "--per_device_reg_head",
    "--patience", str(PATIENCE),
    "--seed", str(SMOKE_SEED),
    "--train_frac", "0.80",
    "--val_frac", "0.10",
    "--sample_frac", "1.0",
    "--num_workers", str(NUM_WORKERS),
    "--log_every", "10",
    "--checkpoint_every", "10",
    "--early_stop_on", _early_stop,
    "--dropout", "0.1",
]

if PHYSICS_WEIGHT > 0:
    _pf_flags = [
        "--pf_data_root", str(PF_DATA_ROOT),
        "--loss_power_balance_weight", str(PHYSICS_WEIGHT),
        "--pf_sparse_y", "1",
        "--pf_huber_delta_kw", "10",
        "--pf_detach_controls",
    ]
    _bal = Path(PF_BALANCE_NODE_LIST_CSV)
    if not _bal.is_absolute():
        _bal = (REPO / _bal).resolve()
    if not _bal.is_file():
        raise FileNotFoundError(f"PF_BALANCE_NODE_LIST_CSV not found: {_bal}")
    _pf_flags.extend(["--pf_balance_node_list_csv", str(_bal)])
    cmd.extend(_pf_flags)

_mode = "physics-informed" if PHYSICS_WEIGHT > 0 else "baseline"
print(f"PHYSICS_WEIGHT={PHYSICS_WEIGHT} ({_mode})")
print(f"early_stop_on={_early_stop}")
print("\nRunning:\n ", " ".join(cmd), "\n", flush=True)

with subprocess.Popen(
    cmd,
    cwd=str(REPO),
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    bufsize=1,
    text=True,
) as proc:
    assert proc.stdout is not None
    for line in proc.stdout:
        print(line, end="")
    rc = proc.wait()

if rc != 0:
    raise subprocess.CalledProcessError(rc, cmd)

print("\nTraining completed.")
print("Run dir:", out_dir.resolve())
print("Checkpoint (best):", (out_dir / "da_gps_multitask_best.pt").resolve())
print("Checkpoint (last):", (out_dir / "training_last.pt").resolve())
print("Regulator classes:", (out_dir / "reg_class_tables.json").resolve())
print("Report:", (out_dir / "da_gps_report.json").resolve())
_run_name = out_dir.name
print("\n=== Section 8 prerequisite ===")
print(f"Baseline run folder:  {out_dir.resolve()}")
print(f"Section 8 expects:    {MYDRIVE_DATA / 'runs' / _run_name}")
print("Checkpoint for init:  da_gps_multitask_best.pt (or training_last.pt)")
if _on_colab() and not str(out_dir.resolve()).startswith(str(MYDRIVE_DATA.resolve())):
    print(
        "WARNING: run dir is NOT under MyDrive/datasets_gnn2 — "
        "copy to Drive before disconnecting the VM or Section 8 will fail."
    )
else:
    print("Checkpoints are on Drive; safe to run Section 8 after this session ends.")
_write_smoke_compare_manifest(
    runs_parent,
    out_dir=out_dir,
    physics_weight=PHYSICS_WEIGHT,
    chunk_glob=CHUNK_GLOB,
    n_chunks=len(chunks),
    epochs=EPOCHS,
    seed=SMOKE_SEED if SMOKE_TEST else 42,
)

````

## 5. Warm-start band daily (non-uniqueness envelope)

For each displayed timestep (`STEP_MIN`), OpenDSS runs **N independent snapshot solves** (`N_WARM_STARTS`, default 10). Each solve randomizes regulator taps and capacitor bank on/off **before** `Solve()` — there is **no** sequential daily QSTS carry-forward between warm starts.

Plots per monitor node:
- **Shaded band** — min/max |V| across the N warm starts
- **Faint lines** — individual warm-start trajectories (optional)
- **Magenta DA-GPS** — single GNN forward per timestep on the same display grid

Use this to check whether DA-GPS |V| lies inside the OpenDSS non-uniqueness envelope.

| Parameter | Meaning |
|-----------|---------|
| `STEP_MIN` | Display/solve resolution in minutes (must divide 60) |
| `N_WARM_STARTS` | Independent random controller inits per timestep |
| `monitor_nodes` | Optional override; default 4 IEEE-8500 candidates |

````python
%matplotlib inline
import sys
for _m in (
    "nonunique_daily_experiment",
    "nonunique_warmstart_band_daily",
    "nonunique_da_gps_daily_compare",
    "nonunique_warmstart_compare",
    "nonunique_four_scenario_demo",
    "nonunique_opendss_daily",
    "nonunique_da_gps",
    "nonunique_plots",
    "run_da_gps_daily_opendss_compare",
):
    sys.modules.pop(_m, None)
from nonunique_daily_experiment import run_and_plot

STEP_MIN = 15          # e.g. 15 -> 96 steps/day; 60 -> 24 steps/day
N_WARM_STARTS = 10     # independent random reg/cap init per timestep

run_and_plot(
    mode="warmstart_band_daily",
    step_min=STEP_MIN,
    n_warm_starts=N_WARM_STARTS,
    include_der=False,
    include_da_gps=True,
    monitor_nodes=["190-8593.1", "190-8581.1", "190-7361.1", "l2973163.2"],
)
````

## 6. Offline physics verification (Windows local)

**Offline check only** — not training. Verify dataset **P_inj/Q_inj** features match surrogate **S(Y, V_label)** on one dailyagg snapshot at label controls/V.

For physics-informed **training**, see **section 3** (`--loss_power_balance_weight`, `--pf_*`).

### Surrogate balance check
- **Y** in Siemens; per-bus kVBase scaling; residuals in **kW/kvar**.
- **r_py** = feature injection − Re/Im(V·conj(YV)) at **label** V.

| Metric | Meaning |
|--------|--------|
| |r_py| P | |P_inj_feature − Re(S(Y,V))| at label V |
| |r_qy| Q | |Q_inj_feature − Im(S(Y,V))| at label V |

Watch **median/p95** on the MV mask, not the max tail. Set `REPO` if auto-detect fails.

USE_ALL_NODES=True (default) evaluates **balance-evaluable** nodes only (known label V on the node and every significant Y-neighbor). False uses the refined MV mask from exclude_interface / hetero_y_neighbors flags. Raw 8531 indexed nodes are **not** balance-evaluable: hetero loads Y-coupled to monitor/cap interface buses (m|p|n) have V=0 on those neighbors, producing MW-scale |r_py| tails while median stays 0. Set SHOW_RAW_INDEXED_DEBUG=True to see that tail.


````python
# --- Offline physics verification (Windows local) ---
# Surrogate Y@V vs dataset P_inj/Q_inj at label V (no OpenDSS).
# Paste into nonunique.ipynb section 6, or run:  python notebook_section6_physics_verify_windows.py
def _in_jupyter() -> bool:
    try:
        get_ipython()  # noqa: F821
        return True
    except NameError:
        return False


_IN_JUPYTER = _in_jupyter()
if _IN_JUPYTER:
    get_ipython().run_line_magic("matplotlib", "inline")  # noqa: F821
else:
    import matplotlib

    matplotlib.use("Agg")

import importlib
import os
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

# === Windows-local paths (edit if your layout differs) ===
REPO = Path(r"C:\Users\alita\OneDrive\Desktop\GNN2")  # or None -> auto-find from cwd
PF_DATA_ROOT = None  # None -> REPO / "datasets_gnn2_from pc/loadtype_8500_dailyagg"
CHUNK_PARENT = None  # optional chunk fallback for CSV discovery

SAMPLE_ID = 0
# True  -> balance-evaluable nodes (known label V on node + all Y-neighbors; default)
# False -> refined MV mask from exclude_interface / hetero_y_neighbors flags below
USE_ALL_NODES = False
SHOW_MASK_COMPARISON = True  # print balance-evaluable vs refined side-by-side
SHOW_RAW_INDEXED_DEBUG = False  # raw 8531-node mask (misleading MW tails; debug only)
USE_RANDOM_CONTROLS = False  # True: small tap noise + random cap flips
EXCLUDE_INTERFACE_BUSES = True       # regxfmr / 190- / m|p|n interface buses (refined mask only)
HETERO_Y_NEIGHBORS_ONLY = True       # hetero load nodes with hetero-only Y neighbors (refined mask only)
# None -> distance-based refined mask (~69 nodes on dailyagg with defaults).
# Set to training list for parity with cell 17 physics loss node set (1177 nodes;
# absolute |r_py| is not meaningful there — flow_relative training ignores P_inj).
PF_BALANCE_NODE_LIST_CSV = None  # e.g. REPO / "colab_pf_data/pf_balance_nodes_explicit.csv"


def _find_repo(explicit: Path | None) -> Path:
    if explicit is not None:
        p = explicit.expanduser().resolve()
        if (p / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
            return p
        raise FileNotFoundError(f"Trainer script not found under REPO={p}")
    for cand in (Path.cwd(), Path.cwd().parent):
        p = cand.resolve()
        if (p / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
            return p
    raise FileNotFoundError(
        "Could not find GNN2 repo. Set REPO at top of cell or cd to repo first."
    )


def _display_or_save(fig, stem: str) -> None:
    if _IN_JUPYTER:
        from IPython.display import display

        display(fig)
    else:
        out = REPO / "outputs" / f"{stem}_sample{SAMPLE_ID}.png"
        out.parent.mkdir(parents=True, exist_ok=True)
        fig.savefig(out, dpi=120, bbox_inches="tight")
        print(f"Saved figure: {out}")
        plt.close(fig)


REPO = _find_repo(REPO)
if PF_DATA_ROOT is None:
    PF_DATA_ROOT = REPO / "datasets_gnn2_from pc" / "loadtype_8500_dailyagg"
else:
    PF_DATA_ROOT = Path(PF_DATA_ROOT).expanduser().resolve()
if CHUNK_PARENT is not None:
    CHUNK_PARENT = Path(CHUNK_PARENT).expanduser().resolve()

if str(REPO) not in sys.path:
    sys.path.insert(0, str(REPO))
os.environ["GNN2_REPO_ROOT"] = str(REPO)  # overwrite stale Colab/content paths

from gnn2_pf_data_paths import resolve_pf_catalog_paths

_, _, PF_DATA_ROOT = resolve_pf_catalog_paths(
    repo=REPO,
    preferred_root=PF_DATA_ROOT if PF_DATA_ROOT.is_dir() else None,
    chunk_parent=CHUNK_PARENT if CHUNK_PARENT is not None and CHUNK_PARENT.is_dir() else None,
)

import gnn2_pf_physics_verify as pf_verify

importlib.reload(pf_verify)

print(f"Repo:              {REPO}")
print(f"PF_DATA_ROOT:      {PF_DATA_ROOT}")
print(f"sample_id:         {SAMPLE_ID}")
print(f"use_all_nodes:     {USE_ALL_NODES}  (balance-evaluable when True)")
print(f"exclude_interface: {EXCLUDE_INTERFACE_BUSES}")
print(f"hetero_y_nbrs:     {HETERO_Y_NEIGHBORS_ONLY}")

snap = pf_verify.load_snapshot_state(
    SAMPLE_ID,
    repo=REPO,
    pf_data_root=PF_DATA_ROOT,
    chunk_parent=CHUNK_PARENT,
    exclude_interface_buses=EXCLUDE_INTERFACE_BUSES,
    hetero_y_neighbors_only=HETERO_Y_NEIGHBORS_ONLY,
    pf_balance_node_list_csv=PF_BALANCE_NODE_LIST_CSV,
)
if USE_RANDOM_CONTROLS:
    rng = np.random.default_rng(42)
    snap = pf_verify.apply_random_perturbation(
        snap, rng=rng, sigma_v_ri=0.0, sigma_tap=0.01, flip_cap_prob=0.15
    )
    print("Applied random control perturbation (label V unchanged)")

if SHOW_MASK_COMPARISON:
    pf_verify.print_verify_mask_comparison(
        snap,
        sample_id=SAMPLE_ID,
        show_raw_indexed=SHOW_RAW_INDEXED_DEBUG,
    )

verify = pf_verify.pytorch_verify_at_state(snap, use_all_nodes=USE_ALL_NODES)
pf_verify.print_pytorch_verify_report(verify, sample_id=SAMPLE_ID)

outlier_tbl = pf_verify.diagnose_pytorch_outliers(snap, verify, top_n=15)
print(f"\n=== Top |dP| outliers ({verify['mask_label']}) ===")
print(outlier_tbl.to_string(index=False))

# --- Residual histograms: |dP| and |dQ| ---
fig_hist, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(
    verify["abs_dp_mv"],
    bins=80,
    log=True,
    color="steelblue",
    alpha=0.85,
)
axes[0].set_title(f"|r_py| at label V ({verify['mask_label']})")
axes[0].set_xlabel("|dP| [kW]")
axes[0].set_ylabel("count")

axes[1].hist(
    verify["abs_dq_mv"],
    bins=80,
    log=True,
    color="darkorange",
    alpha=0.85,
)
axes[1].set_title(f"|r_qy| at label V ({verify['mask_label']})")
axes[1].set_xlabel("|dQ| [kvar]")
axes[1].set_ylabel("count")

plt.tight_layout()
_display_or_save(fig_hist, "pf_verify_hist")

# --- P-Q diagram: feature (P_inj, Q_inj) vs surrogate (P_yv, Q_yv) ---
m = verify["verify_mask"]
p_feat = snap.p_inj_kw[m]
q_feat = snap.q_inj_kvar[m]
p_yv = verify["p_yv_kw"][m]
q_yv = verify["q_yv_kvar"][m]
r_p = verify["dp_kw"][m]
r_q = verify["dq_kvar"][m]

fig_pq, ax = plt.subplots(figsize=(7.5, 7.5))
ax.scatter(p_feat, q_feat, s=12, alpha=0.55, c="steelblue", label="features (P_inj, Q_inj)", zorder=2)
ax.scatter(p_yv, q_yv, s=12, alpha=0.55, c="darkorange", label="surrogate (P_yv, Q_yv)", zorder=3)
ax.quiver(
    p_feat,
    q_feat,
    r_p,
    r_q,
    angles="xy",
    scale_units="xy",
    scale=1.0,
    width=0.002,
    alpha=0.35,
    color="seagreen",
    label="residual (r_p, r_q)",
    zorder=1,
)
ax.axhline(0.0, color="0.7", lw=0.8)
ax.axvline(0.0, color="0.7", lw=0.8)
ax.set_xlabel("P [kW]")
ax.set_ylabel("Q [kvar]")
ax.set_title(f"P-Q plane at label V ({verify['mask_label']})")
ax.legend(loc="best", fontsize=9)
ax.set_aspect("equal", adjustable="box")
plt.tight_layout()
_display_or_save(fig_pq, "pf_verify_pq")
````

## Physics-informed smoke training (cell 17)

Duplicate of section 3 with `PHYSICS_WEIGHT = 0.05`. Uses the same `SMOKE_CHUNK_COUNT`, cache, seed, and architecture as baseline cell 6.

**Prefer cell 19** for baseline+physics+compare in one shot.


````python
# --- Lighter DA-GPS training (long-running; run on Colab GPU or local CUDA) ---
import os
import sys
import subprocess
import datetime
import warnings
from pathlib import Path

DRIVE_ROOT = Path("/content/drive")
MYDRIVE_DATA = DRIVE_ROOT / "MyDrive/datasets_gnn2"
COLAB_CHUNK_DEFAULT = MYDRIVE_DATA / "original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
WIN_CHUNK_DEFAULT = Path(r"D:\datasets\original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")


def _is_windows_drive_path(path) -> bool:
    s = str(path).strip().replace("/", "\\")
    return len(s) >= 2 and s[1] == ":" and s[0].isalpha()


def _on_colab() -> bool:
    return Path("/content").is_dir()


def _drive_mounted() -> bool:
    return DRIVE_ROOT.is_dir() and (DRIVE_ROOT / "MyDrive").is_dir()


def _resolve_data_path(path: Path, *, label: str, colab_fallback: Path | None = None) -> Path:
    """Resolve a data path; never join Windows drive letters with cwd on Linux."""
    raw = str(path)
    if _is_windows_drive_path(raw):
        if _on_colab() and colab_fallback is not None:
            warnings.warn(
                f"{label}={raw!r} is a Windows path on Linux/Colab; using {colab_fallback} instead.",
                UserWarning,
                stacklevel=2,
            )
            return colab_fallback.expanduser().resolve()
        if _on_colab():
            raise ValueError(
                f"{label}={raw!r} is a Windows absolute path and invalid on Colab. "
                f"Mount Drive and use e.g. {COLAB_CHUNK_DEFAULT}"
            )
        return Path(raw).expanduser().resolve()
    p = Path(path).expanduser()
    if p.is_absolute():
        return p.resolve()
    return (Path.cwd() / p).resolve()


def _find_node_pe_csv(chunk_parent: Path) -> Path:
    hits = sorted(chunk_parent.glob("run_*/gnn_node_index_master.csv"))
    if not hits:
        raise FileNotFoundError(
            f"No run_*/gnn_node_index_master.csv under {chunk_parent}"
        )
    return hits[0]


def _find_repo_root() -> Path:
    _env = os.environ.get("GNN2_REPO_ROOT", "").strip()
    candidates = []
    if _env:
        candidates.append(Path(_env).expanduser())
    candidates.extend([
        Path("/content/GNN-Sandia"),
        Path("/content/GNN2"),
        Path.cwd(),
    ])
    seen = set()
    for start in candidates:
        p = start.expanduser().resolve()
        for _ in range(10):
            key = str(p)
            if key not in seen:
                seen.add(key)
                if (p / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
                    return p
            if p.parent == p:
                break
            p = p.parent
    raise FileNotFoundError(
        "Could not find train_da_gps_multitask_complex_voltage_gine.py. "
        "Clone to /content/GNN-Sandia or set REPO = Path('/content/GNN-Sandia')."
    )

def _sorted_run_chunk_dirs(chunk_parent: Path) -> list[Path]:
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_")),
        key=lambda p: p.name,
    )


def _smoke_chunk_subdir_glob(chunk_parent: Path, count: int) -> str:
    """First `count` run_* names, comma-separated for --chunk_subdir_glob."""
    names = [p.name for p in _sorted_run_chunk_dirs(chunk_parent)]
    if len(names) < count:
        raise ValueError(
            f"SMOKE_CHUNK_COUNT={count} but only {len(names)} run_* under {chunk_parent}"
        )
    return ",".join(names[:count])


def _chunks_from_subdir_glob(chunk_parent: Path, glob_pat: str) -> list[Path]:
    import fnmatch

    glob_pat = str(glob_pat).strip()
    if "," in glob_pat:
        allowed = {s.strip() for s in glob_pat.split(",") if s.strip()}
        chunks = sorted(
            (p for p in chunk_parent.iterdir() if p.is_dir() and p.name in allowed),
            key=lambda p: p.name,
        )
        missing = allowed - {p.name for p in chunks}
        if missing:
            raise FileNotFoundError(f"Missing smoke chunk folders: {sorted(missing)}")
        return chunks
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and fnmatch.fnmatch(p.name, glob_pat)),
        key=lambda p: p.name,
    )


def _write_smoke_compare_manifest(
    runs_parent: Path,
    *,
    out_dir: Path,
    physics_weight: float,
    chunk_glob: str,
    n_chunks: int,
    epochs: int,
    seed: int,
) -> None:
    import json as _json

    path = runs_parent / "_smoke_compare_manifest.json"
    manifest = {}
    if path.is_file():
        manifest = _json.loads(path.read_text(encoding="utf-8"))
    key = "physics" if physics_weight > 0 else "baseline"
    manifest[key] = {
        "run_dir": str(out_dir.resolve()),
        "physics_weight": physics_weight,
        "chunk_glob": chunk_glob,
        "n_chunks": n_chunks,
        "epochs": epochs,
        "seed": seed,
    }
    path.write_text(_json.dumps(manifest, indent=2), encoding="utf-8")
    print(f"Smoke compare manifest ({key}) -> {path}")


# Colab: mount Drive when chunks/caches live under MyDrive
if _on_colab() and not _drive_mounted():
    from google.colab import drive
    drive.mount("/content/drive")

# --- smoke / full-run toggles (cells 6 baseline & 17 physics; cell 18 compares) ---
SMOKE_TEST = True
SMOKE_CHUNK_COUNT = 3  # first N run_* alphabetically; 3-5 = fast multi-chunk (not all 40)
SMOKE_EPOCHS = 15      # low for speed; more epochs = stabler MAE, ~linear time
SMOKE_PATIENCE = 5
SMOKE_SEED = 42
SAMPLE_FRAC = 1.0
_DA_CACHE_NAME = "da_gps_chunked_mvagg_smoke_gine" if SMOKE_TEST else "da_gps_chunked_mvagg_full_gine"
# --- paths: auto-detect Colab + Drive; override below if needed ---
REPO = _find_repo_root()

if _on_colab():
    if not _drive_mounted():
        raise RuntimeError(
            "Colab requires Google Drive mounted. Re-run this cell and approve Drive access, "
            "or: from google.colab import drive; drive.mount('/content/drive')"
        )
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    DA_CACHE_ROOT = MYDRIVE_DATA / f"cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = MYDRIVE_DATA / "runs"  # persisted on Drive; /content/GNN-Sandia is ephemeral
elif os.name == "nt":
    CHUNK_PARENT = WIN_CHUNK_DEFAULT
    DA_CACHE_ROOT = REPO / f"datasets_gnn2_from pc/cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
else:
    CHUNK_PARENT = REPO / "datasets_gnn2_from pc/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
    DA_CACHE_ROOT = REPO / f"datasets_gnn2_from pc/cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"

NODE_PE_CSV = None  # auto: first run_*/gnn_node_index_master.csv under CHUNK_PARENT

# --- lighter architecture (sweeps: LIGHTER_LAYERS=1, LIGHTER_HIDDEN=48) ---
LIGHTER_HIDDEN = 64
LIGHTER_LAYERS = 2
LIGHTER_HEADS = 2
LIGHTER_NODE_EMB_DIM = 2

PHYSICS_WEIGHT = 0.05  # 0.0 = baseline; 0.05 = physics-informed (smoke, pu-scaled Huber)
# None -> legacy auto MV mask refinement; set path when physics on
PF_BALANCE_NODE_LIST = REPO / "colab_pf_data/pf_balance_nodes_explicit.csv"

META_AUX_COLS = "pv_pv2_p_post_kw,pv_pv2_q_post_kvar,P_loss_total_post_kw,Q_loss_total_post_kvar"
NUM_WORKERS = 0 if os.name == "nt" else 4

os.chdir(REPO)
sys.path.insert(0, str(REPO))
os.environ["PYTHONUNBUFFERED"] = "1"
os.environ["GNN2_REPO_ROOT"] = str(REPO)

_colab_fb = COLAB_CHUNK_DEFAULT if _on_colab() else None
chunk_parent = _resolve_data_path(CHUNK_PARENT, label="CHUNK_PARENT", colab_fallback=_colab_fb)
if SMOKE_TEST:
    CHUNK_GLOB = _smoke_chunk_subdir_glob(chunk_parent, SMOKE_CHUNK_COUNT)
    EPOCHS = SMOKE_EPOCHS
    PATIENCE = SMOKE_PATIENCE
else:
    CHUNK_GLOB = "run_*"
    EPOCHS = 200
    PATIENCE = 30
da_cache_root = _resolve_data_path(
    DA_CACHE_ROOT,
    label="DA_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / f"cache/{_DA_CACHE_NAME}" if _on_colab() else None,
)
gnn_cache_root = _resolve_data_path(
    GNN_CACHE_ROOT,
    label="GNN_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine" if _on_colab() else None,
)
runs_parent = _resolve_data_path(
    RUNS_PARENT,
    label="RUNS_PARENT",
    colab_fallback=MYDRIVE_DATA / "runs" if _on_colab() else None,
)

from gnn2_pf_data_paths import PF_CAP_NODES_REL, PF_REG_CATALOG_REL, resolve_pf_catalog_paths

PF_DATA_ROOT = None
if PHYSICS_WEIGHT > 0:
    _reg_cat, _cap_map, PF_DATA_ROOT = resolve_pf_catalog_paths(
        repo=REPO,
        preferred_root=None,
        chunk_parent=chunk_parent,
    )

print("=== Preflight ===")
print(f"REPO:           {REPO}")
print(f"SMOKE_TEST:     {SMOKE_TEST}")
print(f"SMOKE_CHUNK_COUNT: {SMOKE_CHUNK_COUNT}")
print(f"CHUNK_GLOB:     {CHUNK_GLOB}")
print(f"SAMPLE_FRAC:    {SAMPLE_FRAC}")
print(f"EPOCHS:         {EPOCHS}")
print(f"PATIENCE:       {PATIENCE}")
print(f"CHUNK_PARENT:   {chunk_parent}")
if not chunk_parent.is_dir():
    raise FileNotFoundError(f"CHUNK_PARENT not found: {chunk_parent}")

run_preview = [p.name for p in _chunks_from_subdir_glob(chunk_parent, CHUNK_GLOB)][:5]
print(f"run_* preview:  {run_preview}")

if NODE_PE_CSV is None:
    node_pe = _find_node_pe_csv(chunk_parent)
else:
    node_pe = _resolve_data_path(NODE_PE_CSV, label="NODE_PE_CSV", colab_fallback=None)
    if not node_pe.is_file():
        warnings.warn(
            f"NODE_PE_CSV not found at {node_pe}; auto-discovering under CHUNK_PARENT.",
            UserWarning,
            stacklevel=2,
        )
        node_pe = _find_node_pe_csv(chunk_parent)

print(f"NODE_PE_CSV:    {node_pe}")
print(f"DA_CACHE_ROOT:  {da_cache_root}")
print(f"GNN_CACHE_ROOT: {gnn_cache_root}")
print(f"RUNS_PARENT:    {runs_parent}")

if PHYSICS_WEIGHT > 0:
    assert PF_DATA_ROOT is not None
    _pf_checks = [
        ("reg_catalog", PF_DATA_ROOT / PF_REG_CATALOG_REL),
        ("cap_nodes", PF_DATA_ROOT / PF_CAP_NODES_REL),
        ("electrical_distance", PF_DATA_ROOT / "electrical_distance_from_substation.csv"),
        (
            "hetero_mv_nodes",
            PF_DATA_ROOT / "Heterogenous GNN dataset/nodes/hetero_mv_nodes_load_transformer.csv",
        ),
        ("bus_kv_cache", PF_DATA_ROOT / "bus_kv_base_by_node.csv"),
    ]
    print(f"PF_DATA_ROOT:   {PF_DATA_ROOT}")
    for label, p in _pf_checks:
        status = "OK" if p.is_file() else "MISSING"
        print(f"  PF {label}: {status}  {p}")
        if not p.is_file():
            raise FileNotFoundError(f"Physics preflight missing {label}: {p}")
    if PF_BALANCE_NODE_LIST is not None:
        _bal = Path(PF_BALANCE_NODE_LIST)
        if not _bal.is_file():
            raise FileNotFoundError(f"PF balance node list not found: {_bal}")
        print(f"PF balance list: OK  {_bal}")
else:
    print("PF_DATA_ROOT:   (skipped - PHYSICS_WEIGHT=0)")

print("=================")

da_cache_root.mkdir(parents=True, exist_ok=True)
gnn_cache_root.mkdir(parents=True, exist_ok=True)
runs_parent.mkdir(parents=True, exist_ok=True)

chunks = _chunks_from_subdir_glob(chunk_parent, CHUNK_GLOB)
if not chunks:
    raise RuntimeError(f"No folders for CHUNK_GLOB={CHUNK_GLOB!r} under {chunk_parent}")

for p in chunks:
    for name in (
        "gnn_node_features_and_targets_mvagg.csv",
        "gnn_edges_phase_static.csv",
        "gnn_sample_meta.csv",
    ):
        if not (p / name).is_file():
            raise FileNotFoundError(f"Missing {name} in {p.name}")

print(f"Found {len(chunks)} chunk(s) matching {CHUNK_GLOB!r} under {chunk_parent}")

tag = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
_pf_suffix = "_pf" if PHYSICS_WEIGHT > 0 else ""
_smoke_suffix = "_smoke" if SMOKE_TEST else ""
out_dir = runs_parent / (
    f"da_gps_chunked_l{LIGHTER_LAYERS}_h{LIGHTER_HIDDEN}_mvagg_gine_metaaux_regce{_pf_suffix}{_smoke_suffix}_{tag}"
)
out_dir.mkdir(parents=True, exist_ok=True)

_early_stop = "voltage" if PHYSICS_WEIGHT > 0 else "total"

cmd = [
    sys.executable, "-u", "train_da_gps_multitask_complex_voltage_gine.py",
    "--chunk_parent", str(chunk_parent),
    "--chunk_subdir_glob", CHUNK_GLOB,
    "--nodes_csv", "gnn_node_features_and_targets_mvagg.csv",
    "--edge_catalog_csv", "gnn_edges_phase_static.csv",
    "--meta_csv", "gnn_sample_meta.csv",
    "--node_feature_cols", "p_load_kw,q_load_kvar,p_pv_kw",
    "--exclude_bess_features",
    "--node_pe_csv", str(node_pe),
    "--node_pe_cols", "auto",
    "--n_system_tokens", "10",
    "--aux_meta_cols", META_AUX_COLS,
    "--lambda_pv", "0.1",
    "--out_dir", str(out_dir),
    "--cache_dir", str(da_cache_root),
    "--bootstrap_gnn_cache_dir", str(gnn_cache_root),
    "--epochs", str(EPOCHS),
    "--batch_size", "64",
    "--hidden", str(LIGHTER_HIDDEN),
    "--layers", str(LIGHTER_LAYERS),
    "--heads", str(LIGHTER_HEADS),
    "--node_emb_dim", str(LIGHTER_NODE_EMB_DIM),
    "--edge_emb_dim", "0",
    "--lr", "5e-4",
    "--weight_decay", "1e-5",
    "--lambda_cap", "0.1",
    "--lambda_reg", "0.1",
    "--reg_loss", "ce",
    "--per_device_cap_head",
    "--per_device_reg_head",
    "--patience", str(PATIENCE),
    "--seed", str(SMOKE_SEED),
    "--train_frac", "0.80",
    "--val_frac", "0.10",
    "--sample_frac", str(SAMPLE_FRAC),
    "--num_workers", str(NUM_WORKERS),
    "--log_every", "10",
    "--checkpoint_every", "10",
    "--early_stop_on", _early_stop,
    "--dropout", "0.1",
]

if PHYSICS_WEIGHT > 0:
    cmd.extend([
        "--pf_data_root", str(PF_DATA_ROOT),
        "--loss_power_balance_weight", str(PHYSICS_WEIGHT),
        "--pf_sparse_y", "1",
        "--pf_huber_delta_kw", "10",
        "--pf_detach_controls",
    ])
    if PF_BALANCE_NODE_LIST is not None:
        cmd.extend(["--pf_balance_node_list_csv", str(PF_BALANCE_NODE_LIST)])

_mode = "physics-informed" if PHYSICS_WEIGHT > 0 else "baseline"
print(f"PHYSICS_WEIGHT={PHYSICS_WEIGHT} ({_mode})")
print(f"early_stop_on={_early_stop}")
print("\nRunning:\n ", " ".join(cmd), "\n", flush=True)

with subprocess.Popen(
    cmd,
    cwd=str(REPO),
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    bufsize=1,
    text=True,
) as proc:
    assert proc.stdout is not None
    for line in proc.stdout:
        print(line, end="")
    rc = proc.wait()

if rc != 0:
    raise subprocess.CalledProcessError(rc, cmd)

print("\nTraining completed.")
print("Run dir:", out_dir.resolve())
print("Checkpoint (best):", (out_dir / "da_gps_multitask_best.pt").resolve())
print("Checkpoint (last):", (out_dir / "training_last.pt").resolve())
print("Regulator classes:", (out_dir / "reg_class_tables.json").resolve())
print("Report:", (out_dir / "da_gps_report.json").resolve())
_run_name = out_dir.name
print("\n=== Section 8 prerequisite ===")
print(f"Baseline run folder:  {out_dir.resolve()}")
print(f"Section 8 expects:    {MYDRIVE_DATA / 'runs' / _run_name}")
print("Checkpoint for init:  da_gps_multitask_best.pt (or training_last.pt)")
if _on_colab() and not str(out_dir.resolve()).startswith(str(MYDRIVE_DATA.resolve())):
    print(
        "WARNING: run dir is NOT under MyDrive/datasets_gnn2 — "
        "copy to Drive before disconnecting the VM or Section 8 will fail."
    )
else:
    print("Checkpoints are on Drive; safe to run Section 8 after this session ends.")
_write_smoke_compare_manifest(
    runs_parent,
    out_dir=out_dir,
    physics_weight=PHYSICS_WEIGHT,
    chunk_glob=CHUNK_GLOB,
    n_chunks=len(chunks),
    epochs=EPOCHS,
    seed=SMOKE_SEED if SMOKE_TEST else 42,
)
````

## 7. Smoke compare (baseline vs physics) — **one cell**

**Run cell 19 only** — baseline then physics-informed training on the same smoke settings.

### Physics loss modes

| Mode | Formula | CLI |
|---|---|---|
| **flow_relative** (default) | `Huber(S(Y,V_label) − S(Y,V_pred))` | `--pf_loss_mode flow_relative` |
| **hybrid** | `L_flow_relative + α·L_absolute` (flow term volt-normalized) | `--pf_loss_mode hybrid --pf_absolute_weight 0.1` |
| **absolute** (legacy) | `Huber(P_inj − S(Y,V_pred))` | `--pf_loss_mode absolute` |

**flow_relative** penalizes network power-flow inconsistency vs the label solution — injection
`P_inj` cancels, so feature injection errors no longer dominate the gradient near `V ≈ V_label`.

**hybrid** adds a small absolute injection-mismatch term (`P_inj = P_pv − P_load`, `Q_inj = −Q_pv − Q_load`)
with configurable `α` (`--pf_absolute_weight`, default 0.1) to nudge toward feature-consistent injections
without reverting to pure absolute loss.

`--pf_auto_scale_volt` optionally scales physics to voltage MSE each batch (ratio capped at
`--pf_auto_scale_max`, default 100). `flow_relative` / hybrid flow term are pre-scaled by `1/mean(y_std²)` so
physics magnitude matches normalized voltage MSE without exploding gradients.

**Node-count weight scaling:** when using the full 1177-node list, pass
`--pf_weight_scale_ref_nodes 185` so effective weight is
`base_weight × (185/N)^0.5` (~0.02 at base 0.05). The refined 185-node list needs no scaling.

**Baseline MAE note:** baseline runs omit all `--pf_*` flags — physics node list does **not**
affect baseline training. With `SMOKE_NO_EARLY_STOP=True`, both runs train the full epoch budget;
test |V| MAE can still differ slightly run-to-run (meta-aux noise).

| Setting | Smoke default |
|---|---|
| Chunks | first `SMOKE_CHUNK_COUNT` (3) `run_*` |
| Epochs | `SMOKE_EPOCHS` (50) — full budget, no early stop |
| Early stop | `SMOKE_NO_EARLY_STOP=True` (`--no_early_stop`; still tracks best val_volt ckpt) |
| Seed | `SMOKE_SEED` (42) |
| Physics | `flow_relative`, weight `0.02`, warmup 3 + ramp 2 epochs, all balance nodes |
| 3-way compare | `SMOKE_RUN_ALL_PF_MODES=True` → train baseline + flow_relative + hybrid (~3× epoch budget) |
| Hybrid-only | `SMOKE_RUN_HYBRID=True`, others `False`, set `SMOKE_REUSE_*_DIR` for prior runs |
| Hybrid α | `SMOKE_PF_ABSOLUTE_WEIGHT=0.1` |
| PF balance nodes | `PF_BALANCE_NODE_LIST` → **`colab_pf_data/pf_balance_nodes_refined.csv`** (**185** Y-neighbor refined; proven smoke winner). Full hetero: `pf_balance_nodes_explicit.csv` (**1177**) + enable `PF_WEIGHT_SCALE_REF_NODES=185`. |
| Hard-example physics | `PF_HARD_NODE_TOPK=0` (all balance nodes; try 48 if physics underfits) |

### Run selection (cell 19)

| Flag | Meaning |
|---|---|
| `SMOKE_RUN_ALL_PF_MODES=True` | Convenience: train all three (baseline + flow_relative + hybrid) |
| `SMOKE_RUN_BASELINE` | Train baseline (`PHYSICS_WEIGHT=0`) |
| `SMOKE_RUN_FLOW_RELATIVE` | Train flow_relative physics |
| `SMOKE_RUN_HYBRID` | Train hybrid physics |
| `SMOKE_REUSE_BASELINE_DIR` | Skip baseline training; load metrics from prior run dir |
| `SMOKE_REUSE_FLOW_DIR` | Skip flow_relative training; load metrics from prior run dir |

When `SMOKE_RUN_ALL_PF_MODES=False`, per-run flags apply. Set reuse dirs to include completed
runs in the compare table without retraining. Paths may be absolute or relative to `RUNS_PARENT`.

**Example — hybrid only, reuse prior Colab smoke runs:**

```python
SMOKE_RUN_ALL_PF_MODES = False
SMOKE_RUN_BASELINE = False
SMOKE_RUN_FLOW_RELATIVE = False
SMOKE_RUN_HYBRID = True
SMOKE_REUSE_BASELINE_DIR = RUNS_PARENT / "da_gps_chunked_l2_h64_mvagg_gine_metaaux_regce_smoke_20260705_054805"
SMOKE_REUSE_FLOW_DIR = RUNS_PARENT / "da_gps_chunked_l2_h64_mvagg_gine_metaaux_regce_pf_w0.02_smoke_20260705_060218"
```

### Making physics more influential (beyond weight tuning)

1. **Refined 185-node list (default)** — offline Y-neighbor + interface filter; expect `masked_nodes=185/3817`.
2. **Full 1177-node list** — broader gradient coverage; enable `--pf_weight_scale_ref_nodes 185` so weight is not diluted.
3. **Full 40-chunk training** — smoke uses 3 chunks; physics regularizes generalization more when the train set is larger.
4. **Physics curriculum** — `PF_WEIGHT_WARMUP_EPOCHS=3`, `PF_WEIGHT_RAMP_EPOCHS=2` (aux-style linear ramp; avoids early val_volt interference).
5. **Label controls (trainer default)** — Y-bus uses ground-truth cap/reg targets, not noisy early predictions.
6. **Hard-example physics** — optional `PF_HARD_NODE_TOPK=48` to focus PF loss on worst voltage errors each batch.

### Compare table (cell 19)

After all runs, cell 19 prints **val and test** columns for the same voltage metrics read from
each run's ``da_gps_report.json`` (best checkpoint):

- **|V| MAE (pu)**
- **Angle MAE (deg)**
- **Norm Re/Im MSE** (training ``val_volt`` loss; normalized Re/Im MSE)

With reuse dirs + `SMOKE_RUN_HYBRID=True`, the table still shows **baseline vs flow_relative vs hybrid**
(three columns) while training only the missing run(s).

Older reports without ``val_metrics`` fall back to ``best_val`` for val Norm Re/Im MSE only; re-run
smoke (or any training) with the updated trainer to populate full val |V|/angle metrics.


````python
# --- Smoke compare: baseline vs physics in ONE cell (Colab GPU or local CUDA) ---
# Runs baseline (PHYSICS_WEIGHT=0) then physics (flow_relative and/or hybrid), then compares.
import json
import os
import sys
import subprocess
import datetime
import warnings
from pathlib import Path

import torch

DRIVE_ROOT = Path("/content/drive")
MYDRIVE_DATA = DRIVE_ROOT / "MyDrive/datasets_gnn2"
COLAB_CHUNK_DEFAULT = MYDRIVE_DATA / "original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
WIN_CHUNK_DEFAULT = Path(r"D:\datasets\original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")


def _is_windows_drive_path(path) -> bool:
    s = str(path).strip().replace("/", "\\")
    return len(s) >= 2 and s[1] == ":" and s[0].isalpha()


def _on_colab() -> bool:
    return Path("/content").is_dir()


def _drive_mounted() -> bool:
    return DRIVE_ROOT.is_dir() and (DRIVE_ROOT / "MyDrive").is_dir()


def _resolve_data_path(path: Path, *, label: str, colab_fallback: Path | None = None) -> Path:
    """Resolve a data path; never join Windows drive letters with cwd on Linux."""
    raw = str(path)
    if _is_windows_drive_path(raw):
        if _on_colab() and colab_fallback is not None:
            warnings.warn(
                f"{label}={raw!r} is a Windows path on Linux/Colab; using {colab_fallback} instead.",
                UserWarning,
                stacklevel=2,
            )
            return colab_fallback.expanduser().resolve()
        if _on_colab():
            raise ValueError(
                f"{label}={raw!r} is a Windows absolute path and invalid on Colab. "
                f"Mount Drive and use e.g. {COLAB_CHUNK_DEFAULT}"
            )
        return Path(raw).expanduser().resolve()
    p = Path(path).expanduser()
    if p.is_absolute():
        return p.resolve()
    return (Path.cwd() / p).resolve()


def _find_node_pe_csv(chunk_parent: Path) -> Path:
    hits = sorted(chunk_parent.glob("run_*/gnn_node_index_master.csv"))
    if not hits:
        raise FileNotFoundError(
            f"No run_*/gnn_node_index_master.csv under {chunk_parent}"
        )
    return hits[0]


def _find_repo_root() -> Path:
    _env = os.environ.get("GNN2_REPO_ROOT", "").strip()
    candidates = []
    if _env:
        candidates.append(Path(_env).expanduser())
    candidates.extend([
        Path("/content/GNN-Sandia"),
        Path("/content/GNN2"),
        Path.cwd(),
    ])
    seen = set()
    for start in candidates:
        p = start.expanduser().resolve()
        for _ in range(10):
            key = str(p)
            if key not in seen:
                seen.add(key)
                if (p / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
                    return p
            if p.parent == p:
                break
            p = p.parent
    raise FileNotFoundError(
        "Could not find train_da_gps_multitask_complex_voltage_gine.py. "
        "Clone to /content/GNN-Sandia or set GNN2_REPO_ROOT."
    )


def _sorted_run_chunk_dirs(chunk_parent: Path) -> list[Path]:
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_")),
        key=lambda p: p.name,
    )


def _smoke_chunk_subdir_glob(chunk_parent: Path, count: int) -> str:
    names = [p.name for p in _sorted_run_chunk_dirs(chunk_parent)]
    if len(names) < count:
        raise ValueError(
            f"SMOKE_CHUNK_COUNT={count} but only {len(names)} run_* under {chunk_parent}"
        )
    return ",".join(names[:count])


def _chunks_from_subdir_glob(chunk_parent: Path, glob_pat: str) -> list[Path]:
    import fnmatch

    glob_pat = str(glob_pat).strip()
    if "," in glob_pat:
        allowed = {s.strip() for s in glob_pat.split(",") if s.strip()}
        chunks = sorted(
            (p for p in chunk_parent.iterdir() if p.is_dir() and p.name in allowed),
            key=lambda p: p.name,
        )
        missing = allowed - {p.name for p in chunks}
        if missing:
            raise FileNotFoundError(f"Missing smoke chunk folders: {sorted(missing)}")
        return chunks
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and fnmatch.fnmatch(p.name, glob_pat)),
        key=lambda p: p.name,
    )


def _write_smoke_compare_manifest(
    runs_parent: Path,
    *,
    out_dir: Path,
    physics_weight: float,
    chunk_glob: str,
    n_chunks: int,
    epochs: int,
    seed: int,
    pf_loss_mode: str = "flow_relative",
    pf_absolute_weight: float = 0.1,
) -> None:
    path = runs_parent / "_smoke_compare_manifest.json"
    manifest = {}
    if path.is_file():
        manifest = json.loads(path.read_text(encoding="utf-8"))
    if physics_weight > 0:
        key = f"physics_{pf_loss_mode}_w{physics_weight:g}"
        manifest.setdefault("physics_runs", [])
        manifest["physics_runs"].append({
            "run_dir": str(out_dir.resolve()),
            "physics_weight": physics_weight,
            "pf_loss_mode": pf_loss_mode,
            "pf_absolute_weight": pf_absolute_weight,
            "chunk_glob": chunk_glob,
            "n_chunks": n_chunks,
            "epochs": epochs,
            "seed": seed,
        })
    else:
        key = "baseline"
        manifest[key] = {
            "run_dir": str(out_dir.resolve()),
            "physics_weight": physics_weight,
            "chunk_glob": chunk_glob,
            "n_chunks": n_chunks,
            "epochs": epochs,
            "seed": seed,
        }
    path.write_text(json.dumps(manifest, indent=2), encoding="utf-8")
    print(f"Smoke compare manifest ({key}) -> {path}")


def _smoke_metrics(run_dir: Path) -> dict:
    """Val/test voltage metrics from da_gps_report.json (best checkpoint)."""
    run_dir = Path(run_dir)
    rep_path = run_dir / "da_gps_report.json"
    if not rep_path.is_file():
        raise FileNotFoundError(f"Missing {rep_path}")
    rep = json.loads(rep_path.read_text(encoding="utf-8"))
    vm = rep.get("val_metrics") or {}
    tm = rep.get("test_metrics") or {}
    hp = rep.get("hyperparameters") or {}
    early_stop_on = hp.get("early_stop_on", "total")
    best_val_early_stop = rep.get("best_val")
    best_val_r2_mean = rep.get("best_val_r2_mean")
    best_val_r2_min = rep.get("best_val_r2_min")
    best_epoch = rep.get("best_epoch")
    last_epoch = rep.get("epochs_completed")
    last_ckpt = run_dir / "training_last.pt"
    if last_ckpt.is_file():
        ck = torch.load(last_ckpt, map_location="cpu", weights_only=False)
        best_val_early_stop = ck.get("best_val", best_val_early_stop)
        best_val_r2_mean = ck.get("best_val_r2_mean", best_val_r2_mean)
        best_val_r2_min = ck.get("best_val_r2_min", best_val_r2_min)
        best_epoch = ck.get("best_epoch", best_epoch)
        last_epoch = ck.get("epoch", last_epoch)
    val_mse_ri = vm.get("mse_ri_normalized")
    if val_mse_ri is None and early_stop_on == "voltage":
        val_mse_ri = best_val_early_stop
    return {
        "run_dir": str(run_dir),
        "n_chunks": rep.get("n_chunks"),
        "val_mae_vmag_pu": vm.get("mae_vmag_pu"),
        "val_mae_angle_deg": vm.get("mae_angle_deg"),
        "val_mse_ri_normalized": val_mse_ri,
        "test_mae_vmag_pu": tm.get("mae_vmag_pu"),
        "test_mae_angle_deg": tm.get("mae_angle_deg"),
        "test_mse_ri_normalized": tm.get("mse_ri_normalized"),
        "test_r2_mean": tm.get("r2_vmag_mean"),
        "test_r2_min": tm.get("r2_vmag_min"),
        "best_val_early_stop": best_val_early_stop,
        "best_val_volt": val_mse_ri,
        "best_val_r2_mean": best_val_r2_mean,
        "best_val_r2_min": best_val_r2_min,
        "best_epoch": best_epoch,
        "early_stop_on": early_stop_on,
        "physics_weight": float(hp.get("loss_power_balance_weight") or 0.0),
        "pf_loss_mode": hp.get("pf_loss_mode") or "flow_relative",
        "pf_absolute_weight": float(hp.get("pf_absolute_weight") or 0.0),
        "chunk_glob": hp.get("chunk_subdir_glob"),
        "epochs": hp.get("epochs"),
        "epochs_completed": last_epoch,
        "no_early_stop": bool(hp.get("no_early_stop")),
    }


def _fmt_metric(x, *, width: int = 14, prec: int = 6) -> str:
    if x is None:
        return f"{'n/a':>{width}}"
    fmt = f"{{:>{width}.{prec}f}}"
    return fmt.format(float(x))


def _fmt_delta(a, b, *, width: int = 14, prec: int = 6) -> str:
    if a is None or b is None:
        return f"{'n/a':>{width}}"
    fmt = f"{{:>{width}.{prec}f}}"
    return fmt.format(float(b) - float(a))




def _print_smoke_multi_row(
    label: str,
    key: str,
    runs: list[tuple[str, dict]],
    *,
    prec: int = 6,
) -> None:
    """Metric row: val/test for each named run (baseline first)."""
    parts = [f"{label:18}"]
    for _name, m in runs:
        parts.append(_fmt_metric(m.get(f"val_{key}"), prec=prec))
    parts.append("|")
    for _name, m in runs:
        parts.append(_fmt_metric(m.get(f"test_{key}"), prec=prec))
    print(" ".join(parts))


def _best_physics_by_test_mae(physics_metrics: list[tuple[str, dict]]) -> tuple[str, dict]:
    best_name, best_m = physics_metrics[0]
    for name, m in physics_metrics[1:]:
        if m.get("test_mae_vmag_pu") is not None and (
            best_m.get("test_mae_vmag_pu") is None
            or float(m["test_mae_vmag_pu"]) < float(best_m["test_mae_vmag_pu"])
        ):
            best_name, best_m = name, m
    return best_name, best_m



def _resolve_smoke_run_dir(path: Path | str | None, *, label: str, runs_parent: Path) -> Path:
    """Resolve a prior smoke run dir and verify da_gps_report.json exists."""
    if path is None:
        raise ValueError(f"{label}: reuse path is None")
    p = Path(path).expanduser()
    if not p.is_absolute():
        p = (runs_parent / p).resolve()
    else:
        p = p.resolve()
    rep = p / "da_gps_report.json"
    if not rep.is_file():
        raise FileNotFoundError(f"{label}: missing {rep}")
    return p


def _smoke_metrics_from_reuse(path: Path | str | None, *, label: str, runs_parent: Path) -> dict:
    rdir = _resolve_smoke_run_dir(path, label=label, runs_parent=runs_parent)
    print(f"Reusing {label} metrics from {rdir} (skip training)")
    return _smoke_metrics(rdir)

def _print_smoke_metric_row(label: str, key: str, base: dict, phys: dict, *, prec: int = 6) -> None:
    """One metric row with aligned val/test columns (baseline, physics, delta)."""
    print(
        f"{label:18} "
        f"{_fmt_metric(base.get(f'val_{key}'), prec=prec)} "
        f"{_fmt_metric(phys.get(f'val_{key}'), prec=prec)} "
        f"{_fmt_delta(base.get(f'val_{key}'), phys.get(f'val_{key}'), prec=prec)} | "
        f"{_fmt_metric(base.get(f'test_{key}'), prec=prec)} "
        f"{_fmt_metric(phys.get(f'test_{key}'), prec=prec)} "
        f"{_fmt_delta(base.get(f'test_{key}'), phys.get(f'test_{key}'), prec=prec)}"
    )


def _pf_preflight(pf_data_root: Path) -> None:
    from gnn2_pf_data_paths import PF_CAP_NODES_REL, PF_REG_CATALOG_REL

    _pf_checks = [
        ("reg_catalog", pf_data_root / PF_REG_CATALOG_REL),
        ("cap_nodes", pf_data_root / PF_CAP_NODES_REL),
        ("electrical_distance", pf_data_root / "electrical_distance_from_substation.csv"),
        (
            "hetero_mv_nodes",
            pf_data_root / "Heterogenous GNN dataset/nodes/hetero_mv_nodes_load_transformer.csv",
        ),
        ("bus_kv_cache", pf_data_root / "bus_kv_base_by_node.csv"),
    ]
    print(f"PF_DATA_ROOT:   {pf_data_root}")
    for label, p in _pf_checks:
        status = "OK" if p.is_file() else "MISSING"
        print(f"  PF {label}: {status}  {p}")
        if not p.is_file():
            raise FileNotFoundError(f"Physics preflight missing {label}: {p}")


def _pf_balance_node_count(list_csv: Path | None) -> int | None:
    if list_csv is None or not Path(list_csv).is_file():
        return None
    import pandas as pd
    df = pd.read_csv(list_csv)
    if "node" in df.columns:
        return int(df["node"].dropna().nunique())
    if "node_idx" in df.columns:
        return int(df["node_idx"].dropna().nunique())
    return len(df)




def _trainer_supports_no_early_stop(repo: Path) -> bool:
    """True when train_da_gps_multitask_complex_voltage_gine.py accepts --no_early_stop."""
    trainer = repo / "train_da_gps_multitask_complex_voltage_gine.py"
    if not trainer.is_file():
        return False
    try:
        proc = subprocess.run(
            [sys.executable, str(trainer), "--help"],
            capture_output=True,
            text=True,
            timeout=120,
        )
    except (OSError, subprocess.TimeoutExpired):
        return False
    help_text = (proc.stdout or "") + (proc.stderr or "")
    return "--no_early_stop" in help_text

def run_smoke_training(
    *,
    physics_weight: float,
    repo: Path,
    chunk_parent: Path,
    chunk_glob: str,
    chunks: list[Path],
    node_pe: Path,
    da_cache_root: Path,
    gnn_cache_root: Path,
    runs_parent: Path,
    epochs: int,
    patience: int,
    seed: int,
    sample_frac: float,
    pf_balance_node_list: Path | None,
    pf_loss_mode: str = "flow_relative",
    pf_absolute_weight: float = 0.1,
) -> Path:
    from gnn2_pf_data_paths import resolve_pf_catalog_paths

    if SMOKE_NO_EARLY_STOP and not _trainer_supports_no_early_stop(repo):
        patience = max(int(patience), int(epochs) + 1)

    pf_data_root = None
    if physics_weight > 0:
        _reg_cat, _cap_map, pf_data_root = resolve_pf_catalog_paths(
            repo=repo,
            preferred_root=None,
            chunk_parent=chunk_parent,
        )
        _pf_preflight(pf_data_root)
        if pf_balance_node_list is not None and not pf_balance_node_list.is_file():
            raise FileNotFoundError(f"PF balance node list not found: {pf_balance_node_list}")
        if pf_balance_node_list is not None:
            _n_bal = _pf_balance_node_count(pf_balance_node_list)
            print(f"PF balance list: OK  {pf_balance_node_list}  (n_nodes={_n_bal})")

    tag = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
    if physics_weight > 0:
        _pf_suffix = f"_pf_{pf_loss_mode}_w{physics_weight:g}"
    else:
        _pf_suffix = ""
    out_dir = runs_parent / (
        f"da_gps_chunked_l{LIGHTER_LAYERS}_h{LIGHTER_HIDDEN}_mvagg_gine_metaaux_regce{_pf_suffix}_smoke_{tag}"
    )
    out_dir.mkdir(parents=True, exist_ok=True)

    # Fair compare: both baseline and physics early-stop on val_volt (not mixed total vs voltage).
    _early_stop = "voltage"
    _mode = "physics-informed" if physics_weight > 0 else "baseline"
    print(f"\n{'=' * 60}")
    print(f"SMOKE RUN: PHYSICS_WEIGHT={physics_weight} ({_mode})  pf_loss_mode={pf_loss_mode}")
    print(f"early_stop_on={_early_stop}")
    print(f"out_dir={out_dir}")
    print(f"{'=' * 60}")

    cmd = [
        sys.executable, "-u", "train_da_gps_multitask_complex_voltage_gine.py",
        "--chunk_parent", str(chunk_parent),
        "--chunk_subdir_glob", chunk_glob,
        "--nodes_csv", "gnn_node_features_and_targets_mvagg.csv",
        "--edge_catalog_csv", "gnn_edges_phase_static.csv",
        "--meta_csv", "gnn_sample_meta.csv",
        "--node_feature_cols", "p_load_kw,q_load_kvar,p_pv_kw",
        "--exclude_bess_features",
        "--node_pe_csv", str(node_pe),
        "--node_pe_cols", "auto",
        "--n_system_tokens", "10",
        "--aux_meta_cols", META_AUX_COLS,
        "--lambda_pv", "0.1",
        "--out_dir", str(out_dir),
        "--cache_dir", str(da_cache_root),
        "--bootstrap_gnn_cache_dir", str(gnn_cache_root),
        "--epochs", str(epochs),
        "--batch_size", "64",
        "--hidden", str(LIGHTER_HIDDEN),
        "--layers", str(LIGHTER_LAYERS),
        "--heads", str(LIGHTER_HEADS),
        "--node_emb_dim", str(LIGHTER_NODE_EMB_DIM),
        "--edge_emb_dim", "0",
        "--lr", "5e-4",
        "--weight_decay", "1e-5",
        "--lambda_cap", "0.1",
        "--lambda_reg", "0.1",
        "--reg_loss", "ce",
        "--per_device_cap_head",
        "--per_device_reg_head",
        "--patience", str(patience),
        "--seed", str(seed),
        "--train_frac", "0.80",
        "--val_frac", "0.10",
        "--sample_frac", str(sample_frac),
        "--num_workers", str(NUM_WORKERS),
        "--log_every", "10",
        "--checkpoint_every", "10",
        "--early_stop_on", _early_stop,
        "--dropout", "0.1",
    ]
    if SMOKE_NO_EARLY_STOP:
        if _trainer_supports_no_early_stop(repo):
            cmd.append("--no_early_stop")
        else:
            warnings.warn(
                "Trainer lacks --no_early_stop; using patience="
                f"{max(int(patience), int(epochs) + 1)} instead (git pull repo).",
                stacklevel=2,
            )

    if physics_weight > 0:
        cmd.extend([
            "--pf_data_root", str(pf_data_root),
            "--loss_power_balance_weight", str(physics_weight),
            "--pf_loss_mode", pf_loss_mode,
            "--pf_sparse_y", "1",
            "--pf_huber_delta_kw", "10",
            "--pf_detach_controls",
        ])
        if PF_AUTO_SCALE_VOLT:
            cmd.extend(["--pf_auto_scale_volt", "--pf_auto_scale_max", "100"])
        if PF_WEIGHT_SCALE_REF_NODES and float(PF_WEIGHT_SCALE_REF_NODES) > 0:
            cmd.extend([
                "--pf_weight_scale_ref_nodes", str(PF_WEIGHT_SCALE_REF_NODES),
                "--pf_weight_scale_alpha", str(PF_WEIGHT_SCALE_ALPHA),
            ])
        if PF_HARD_NODE_TOPK and int(PF_HARD_NODE_TOPK) > 0:
            cmd.extend(["--pf_hard_node_topk", str(int(PF_HARD_NODE_TOPK))])
        if PF_WEIGHT_WARMUP_EPOCHS and int(PF_WEIGHT_WARMUP_EPOCHS) > 0:
            cmd.extend(["--pf_weight_warmup_epochs", str(int(PF_WEIGHT_WARMUP_EPOCHS))])
        if PF_WEIGHT_RAMP_EPOCHS and int(PF_WEIGHT_RAMP_EPOCHS) > 0:
            cmd.extend(["--pf_weight_ramp_epochs", str(int(PF_WEIGHT_RAMP_EPOCHS))])
        if pf_balance_node_list is not None:
            cmd.extend(["--pf_balance_node_list_csv", str(pf_balance_node_list)])
        if pf_loss_mode == "hybrid":
            cmd.extend(["--pf_absolute_weight", str(pf_absolute_weight)])

    print("\nRunning:\n ", " ".join(cmd), "\n", flush=True)

    with subprocess.Popen(
        cmd,
        cwd=str(repo),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        bufsize=1,
        text=True,
    ) as proc:
        assert proc.stdout is not None
        for line in proc.stdout:
            print(line, end="")
        rc = proc.wait()

    if rc != 0:
        raise subprocess.CalledProcessError(rc, cmd)

    print("\nTraining completed.")
    print("Run dir:", out_dir.resolve())
    print("Report:", (out_dir / "da_gps_report.json").resolve())
    _write_smoke_compare_manifest(
        runs_parent,
        out_dir=out_dir,
        physics_weight=physics_weight,
        chunk_glob=chunk_glob,
        n_chunks=len(chunks),
        epochs=epochs,
        seed=seed,
        pf_loss_mode=pf_loss_mode,
        pf_absolute_weight=pf_absolute_weight,
    )
    return out_dir


# --- smoke settings (shared by baseline + physics) ---
SMOKE_CHUNK_COUNT = 3
SMOKE_EPOCHS = 50      # full 50-epoch runs for fair baseline vs physics compare
SMOKE_PATIENCE = 10    # ignored when SMOKE_NO_EARLY_STOP=True
SMOKE_NO_EARLY_STOP = True  # both runs train all epochs; best ckpt still by val_volt
SMOKE_SEED = 42
SAMPLE_FRAC = 1.0
# flow_relative: trainer scales physics to normalized-voltage MSE; fixed weight ~0.01-0.1
SMOKE_PHYSICS_WEIGHT = 0.02  # lighter coupling; flow_relative is volt-normalized in trainer
SMOKE_RUN_ALL_PF_MODES = False  # True = train baseline + flow_relative + hybrid (~3x time)
# Per-run toggles (used when SMOKE_RUN_ALL_PF_MODES=False):
SMOKE_RUN_BASELINE = False
SMOKE_RUN_FLOW_RELATIVE = False
SMOKE_RUN_HYBRID = True
# Reuse prior runs (skip training; load da_gps_report.json). Path = absolute or under RUNS_PARENT.
# Example Colab dirs (fill in after prior smoke runs):
# SMOKE_REUSE_BASELINE_DIR = RUNS_PARENT / "da_gps_chunked_l2_h64_mvagg_gine_metaaux_regce_smoke_20260705_054805"
# SMOKE_REUSE_FLOW_DIR = RUNS_PARENT / "da_gps_chunked_l2_h64_mvagg_gine_metaaux_regce_pf_w0.02_smoke_20260705_060218"
SMOKE_REUSE_BASELINE_DIR = None
SMOKE_REUSE_FLOW_DIR = None
SMOKE_PF_ABSOLUTE_WEIGHT = 0.1  # hybrid alpha (--pf_absolute_weight)
SMOKE_PF_LOSS_MODE = "flow_relative"  # legacy single-mode when only one physics flag is True
PF_AUTO_SCALE_VOLT = False  # optional; ratio capped at 100 when True
PF_WEIGHT_SCALE_REF_NODES = 0  # set 185 when using explicit 1177-node list; 0=off (refined default)
PF_WEIGHT_SCALE_ALPHA = 0.5  # effective_w = base_w * (N_ref/N)^alpha
PF_WEIGHT_WARMUP_EPOCHS = 3  # zero PF ep 1-3 so voltage head stabilizes first
PF_WEIGHT_RAMP_EPOCHS = 2  # linear ramp ep 4-5, full weight from ep 6
PF_HARD_NODE_TOPK = 0  # all balance nodes; top-k can hurt short smoke runs

LIGHTER_HIDDEN = 64
LIGHTER_LAYERS = 2
LIGHTER_HEADS = 2
LIGHTER_NODE_EMB_DIM = 2

META_AUX_COLS = "pv_pv2_p_post_kw,pv_pv2_q_post_kvar,P_loss_total_post_kw,Q_loss_total_post_kvar"
NUM_WORKERS = 0 if os.name == "nt" else 4
_DA_CACHE_NAME = "da_gps_chunked_mvagg_smoke_gine"

# Colab: mount Drive when chunks/caches live under MyDrive
if _on_colab() and not _drive_mounted():
    from google.colab import drive
    drive.mount("/content/drive")

REPO = _find_repo_root()
os.chdir(REPO)
sys.path.insert(0, str(REPO))
os.environ["PYTHONUNBUFFERED"] = "1"
os.environ["GNN2_REPO_ROOT"] = str(REPO)

if _on_colab():
    if not _drive_mounted():
        raise RuntimeError(
            "Colab requires Google Drive mounted. Re-run this cell and approve Drive access, "
            "or: from google.colab import drive; drive.mount('/content/drive')"
        )
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    DA_CACHE_ROOT = MYDRIVE_DATA / f"cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = MYDRIVE_DATA / "runs"  # persisted on Drive; /content/GNN-Sandia is ephemeral
elif os.name == "nt":
    CHUNK_PARENT = WIN_CHUNK_DEFAULT
    DA_CACHE_ROOT = REPO / f"datasets_gnn2_from pc/cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
else:
    CHUNK_PARENT = REPO / "datasets_gnn2_from pc/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
    DA_CACHE_ROOT = REPO / f"datasets_gnn2_from pc/cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"

NODE_PE_CSV = None
PF_BALANCE_NODE_LIST = REPO / "colab_pf_data/pf_balance_nodes_refined.csv"
# Full hetero (~1177): REPO / "colab_pf_data/pf_balance_nodes_explicit.csv"

_colab_fb = COLAB_CHUNK_DEFAULT if _on_colab() else None
chunk_parent = _resolve_data_path(CHUNK_PARENT, label="CHUNK_PARENT", colab_fallback=_colab_fb)
CHUNK_GLOB = _smoke_chunk_subdir_glob(chunk_parent, SMOKE_CHUNK_COUNT)
da_cache_root = _resolve_data_path(
    DA_CACHE_ROOT,
    label="DA_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / f"cache/{_DA_CACHE_NAME}" if _on_colab() else None,
)
gnn_cache_root = _resolve_data_path(
    GNN_CACHE_ROOT,
    label="GNN_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine" if _on_colab() else None,
)
runs_parent = _resolve_data_path(
    RUNS_PARENT,
    label="RUNS_PARENT",
    colab_fallback=MYDRIVE_DATA / "runs" if _on_colab() else None,
)

print("=== Preflight (unified smoke compare) ===")
print(f"REPO:              {REPO}")
print(f"SMOKE_CHUNK_COUNT: {SMOKE_CHUNK_COUNT}")
print(f"SMOKE_EPOCHS:      {SMOKE_EPOCHS}")
print(f"SMOKE_NO_EARLY_STOP: {SMOKE_NO_EARLY_STOP}")
if SMOKE_NO_EARLY_STOP:
    if _trainer_supports_no_early_stop(REPO):
        print(f"early stopping:    OFF (full {SMOKE_EPOCHS} epochs, --no_early_stop)")
    else:
        _pat = max(SMOKE_PATIENCE, SMOKE_EPOCHS + 1)
        print(
            f"early stopping:    OFF via patience={_pat} "
            "(stale trainer; git pull for --no_early_stop)"
        )
else:
    print(f"SMOKE_PATIENCE:    {SMOKE_PATIENCE}")
print(f"SMOKE_SEED:        {SMOKE_SEED}")
print(f"CHUNK_GLOB:        {CHUNK_GLOB}")
print(f"CHUNK_PARENT:      {chunk_parent}")
print(f"early_stop_on:     voltage (both baseline and physics)")
print(f"SMOKE_RUN_ALL_PF_MODES: {SMOKE_RUN_ALL_PF_MODES}")
if SMOKE_RUN_ALL_PF_MODES:
    print("  -> train: baseline + flow_relative + hybrid")
else:
    print(f"SMOKE_RUN_BASELINE: {SMOKE_RUN_BASELINE}  reuse={SMOKE_REUSE_BASELINE_DIR}")
    print(f"SMOKE_RUN_FLOW_RELATIVE: {SMOKE_RUN_FLOW_RELATIVE}  reuse={SMOKE_REUSE_FLOW_DIR}")
    print(f"SMOKE_RUN_HYBRID: {SMOKE_RUN_HYBRID}")
print(f"SMOKE_PF_LOSS_MODE: {SMOKE_PF_LOSS_MODE}")
print(f"SMOKE_PF_ABSOLUTE_WEIGHT: {SMOKE_PF_ABSOLUTE_WEIGHT}")
print(f"PF_AUTO_SCALE_VOLT: {PF_AUTO_SCALE_VOLT}")
print(f"SMOKE_PHYSICS_WEIGHT: {SMOKE_PHYSICS_WEIGHT}")
print(f"PF_BALANCE_NODE_LIST: {PF_BALANCE_NODE_LIST}")
_pf_n = _pf_balance_node_count(PF_BALANCE_NODE_LIST) if PF_BALANCE_NODE_LIST else None
print(f"PF balance nodes (preflight): {_pf_n if _pf_n is not None else 'legacy mv mask'}")
print(f"PF_WEIGHT_SCALE_REF_NODES: {PF_WEIGHT_SCALE_REF_NODES}")
print(f"PF_HARD_NODE_TOPK: {PF_HARD_NODE_TOPK}")
print(f"PF_WEIGHT_WARMUP_EPOCHS: {PF_WEIGHT_WARMUP_EPOCHS}")
print(f"PF_WEIGHT_RAMP_EPOCHS: {PF_WEIGHT_RAMP_EPOCHS}")
print("Note: baseline run omits pf flags — baseline MAE varies ~1e-3 run-to-run independently of physics node list.")
if not chunk_parent.is_dir():
    raise FileNotFoundError(f"CHUNK_PARENT not found: {chunk_parent}")

if NODE_PE_CSV is None:
    node_pe = _find_node_pe_csv(chunk_parent)
else:
    node_pe = _resolve_data_path(NODE_PE_CSV, label="NODE_PE_CSV", colab_fallback=None)
    if not node_pe.is_file():
        node_pe = _find_node_pe_csv(chunk_parent)

print(f"NODE_PE_CSV:       {node_pe}")
print(f"DA_CACHE_ROOT:     {da_cache_root}")
print(f"GNN_CACHE_ROOT:    {gnn_cache_root}")
print(f"RUNS_PARENT:       {runs_parent}")
print("=======================================")

da_cache_root.mkdir(parents=True, exist_ok=True)
gnn_cache_root.mkdir(parents=True, exist_ok=True)
runs_parent.mkdir(parents=True, exist_ok=True)

chunks = _chunks_from_subdir_glob(chunk_parent, CHUNK_GLOB)
if not chunks:
    raise RuntimeError(f"No folders for CHUNK_GLOB={CHUNK_GLOB!r} under {chunk_parent}")

for p in chunks:
    for name in (
        "gnn_node_features_and_targets_mvagg.csv",
        "gnn_edges_phase_static.csv",
        "gnn_sample_meta.csv",
    ):
        if not (p / name).is_file():
            raise FileNotFoundError(f"Missing {name} in {p.name}")

print(f"Found {len(chunks)} chunk(s) for smoke under {chunk_parent}")

_common = dict(
    repo=REPO,
    chunk_parent=chunk_parent,
    chunk_glob=CHUNK_GLOB,
    chunks=chunks,
    node_pe=node_pe,
    da_cache_root=da_cache_root,
    gnn_cache_root=gnn_cache_root,
    runs_parent=runs_parent,
    epochs=SMOKE_EPOCHS,
    patience=SMOKE_PATIENCE,
    seed=SMOKE_SEED,
    sample_frac=SAMPLE_FRAC,
)

if SMOKE_RUN_ALL_PF_MODES:
    _run_baseline = _run_flow = _run_hybrid = True
else:
    _run_baseline = bool(SMOKE_RUN_BASELINE)
    _run_flow = bool(SMOKE_RUN_FLOW_RELATIVE)
    _run_hybrid = bool(SMOKE_RUN_HYBRID)

all_runs: list[tuple[str, dict]] = []
_n_train = 0

if _run_baseline:
    _n_train += 1
    baseline_dir = run_smoke_training(
        physics_weight=0.0,
        pf_balance_node_list=None,
        **_common,
    )
    all_runs.append(("baseline", _smoke_metrics(baseline_dir)))
elif SMOKE_REUSE_BASELINE_DIR is not None:
    all_runs.append((
        "baseline",
        _smoke_metrics_from_reuse(
            SMOKE_REUSE_BASELINE_DIR, label="baseline", runs_parent=runs_parent
        ),
    ))

if _run_flow:
    _n_train += 1
    flow_dir = run_smoke_training(
        physics_weight=SMOKE_PHYSICS_WEIGHT,
        pf_balance_node_list=PF_BALANCE_NODE_LIST,
        pf_loss_mode="flow_relative",
        pf_absolute_weight=SMOKE_PF_ABSOLUTE_WEIGHT,
        **_common,
    )
    all_runs.append(("flow_relative", _smoke_metrics(flow_dir)))
elif SMOKE_REUSE_FLOW_DIR is not None:
    all_runs.append((
        "flow_relative",
        _smoke_metrics_from_reuse(
            SMOKE_REUSE_FLOW_DIR, label="flow_relative", runs_parent=runs_parent
        ),
    ))

if _run_hybrid:
    _n_train += 1
    hybrid_dir = run_smoke_training(
        physics_weight=SMOKE_PHYSICS_WEIGHT,
        pf_balance_node_list=PF_BALANCE_NODE_LIST,
        pf_loss_mode="hybrid",
        pf_absolute_weight=SMOKE_PF_ABSOLUTE_WEIGHT,
        **_common,
    )
    all_runs.append(("hybrid", _smoke_metrics(hybrid_dir)))

if not all_runs:
    raise RuntimeError(
        "No smoke runs configured: enable SMOKE_RUN_* flags or set SMOKE_REUSE_*_DIR"
    )

base = next((m for n, m in all_runs if n == "baseline"), None)
physics_metrics = [(n, m) for n, m in all_runs if n != "baseline"]
_pf_modes = tuple(n for n, _ in physics_metrics)

print("\n" + "=" * 60)
print("=== Smoke compare (full epochs, early_stop_on=voltage) ===")
print(
    f"Compare columns: {', '.join(n for n, _ in all_runs)}  "
    f"(training {_n_train} run(s))  auto_scale_volt={PF_AUTO_SCALE_VOLT}  "
    f"weight={SMOKE_PHYSICS_WEIGHT}  hybrid_alpha={SMOKE_PF_ABSOLUTE_WEIGHT}  balance_nodes={_pf_n}"
)
_pf_full_ep = int(PF_WEIGHT_WARMUP_EPOCHS or 0) + int(PF_WEIGHT_RAMP_EPOCHS or 0)
if _pf_full_ep > 0:
    print(
        f"Physics curriculum: zero PF ep 1..{PF_WEIGHT_WARMUP_EPOCHS}, "
        f"ramp ep {int(PF_WEIGHT_WARMUP_EPOCHS)+1}..{_pf_full_ep} (full weight from ep {_pf_full_ep+1}); "
        "baseline has no PF delay — compare test MAE, not epoch count."
    )
else:
    print("Physics weight: full from epoch 1 (no warmup/ramp)")
print(f"Epoch budget: {SMOKE_EPOCHS} epochs x {_n_train} training run(s)")
if base is not None:
    print(f"Chunks: baseline n={base['n_chunks']}  glob={base.get('chunk_glob')}")
elif all_runs:
    _ref = all_runs[0][1]
    print(f"Chunks: n={_ref.get('n_chunks')}  glob={_ref.get('chunk_glob')}")
for name, m in physics_metrics:
    print(
        f"  {name}: epochs {m.get('epochs_completed') or m.get('epochs') or '?'} "
        f"/ {m.get('epochs') or '?'}  alpha={m.get('pf_absolute_weight')}"
    )
print()
_hdr = f"{'':18} " + " ".join(f"{'val ' + n:>14}" for n, _ in all_runs) + " | " + " ".join(
    f"{'test ' + n:>14}" for n, _ in all_runs
)
print(_hdr)
_print_smoke_multi_row("|V| MAE (pu)", "mae_vmag_pu", all_runs)
_print_smoke_multi_row("Angle MAE (deg)", "mae_angle_deg", all_runs)
_print_smoke_multi_row("Norm Re/Im MSE", "mse_ri_normalized", all_runs)
print()
best_name, best_phys = _best_physics_by_test_mae(physics_metrics)
if (
    base is not None
    and best_phys.get("test_mae_vmag_pu") is not None
    and base.get("test_mae_vmag_pu") is not None
):
    if float(best_phys["test_mae_vmag_pu"]) < float(base["test_mae_vmag_pu"]):
        print(
            f"WINNER (MAE): physics {best_name} "
            f"test |V| MAE {best_phys['test_mae_vmag_pu']:.6f} < baseline {base['test_mae_vmag_pu']:.6f}"
        )
    else:
        print(
            f"WINNER (MAE): baseline "
            f"test |V| MAE {base['test_mae_vmag_pu']:.6f} <= best physics {best_phys['test_mae_vmag_pu']:.6f} ({best_name})"
        )
        print("Tip: try PF_HARD_NODE_TOPK=48, tune hybrid alpha, or more SMOKE_EPOCHS.")
if len(physics_metrics) > 1:
    _flow = next((m for n, m in physics_metrics if n == "flow_relative"), None)
    _hyb = next((m for n, m in physics_metrics if n == "hybrid"), None)
    if _flow and _hyb and _flow.get("test_mae_vmag_pu") is not None and _hyb.get("test_mae_vmag_pu") is not None:
        d = float(_hyb["test_mae_vmag_pu"]) - float(_flow["test_mae_vmag_pu"])
        print(
            f"Hybrid vs flow_relative test |V| MAE delta: {d:+.6f} "
            f"(negative => hybrid better)"
        )
if base is not None:
    print(f"Baseline run: {base['run_dir']}")
for name, m in physics_metrics:
    print(f"Physics ({name}): {m['run_dir']}")
print("=" * 60)
````

## Section 8 — Physics fine-tune from saved baseline (Engage-style)

**Prerequisite:** Run **Section 3** baseline training first (`PHYSICS_WEIGHT=0`, `SMOKE_TEST=False` for the full 40-chunk / 200-epoch run) with **Google Drive mounted** so checkpoints land under `/content/drive/MyDrive/datasets_gnn2/runs/`. Section 8 cannot recover a baseline written only to the ephemeral clone (`/content/GNN-Sandia/...`).

Post-hoc fine-tune a **pretrained baseline** (`PHYSICS_WEIGHT=0`) with a small physics term and optional voltage anchor.

1. **Baseline eval** (`--eval_only`): load `INIT_CHECKPOINT`, report val/test metrics without training.
2. **Fine-tune**: `--init_checkpoint` + physics on full chunks; `--finetune_physics_only` zeros cap/reg/meta-aux losses. Trainer prints `=== Baseline before post-tuning (init checkpoint) ===` with full val/test metrics (|V| MAE, angle MAE, Re/Im MSE, r2_mean, r2_min, worst_mae, tot, volt, pf when physics is on), then `[vs baseline] epoch N …` deltas for every stored metric each epoch, and a comprehensive before/after table at the end.
3. **Save gate** (`--save_only_if_improved`): write `da_gps_multitask_best.pt` only if test |V| MAE improves and angle / norm Re-Im MSE do not regress.

Uses the same `CHUNK_PARENT`, DA cache, architecture as the init run (default: **h=96, L=2**, `da_gps_chunked_l4_mvagg_gine_metaaux_regce_20260516_225149_CCE`), and `reg_class_tables.json` from the init run.
PF balance nodes: refined **185**-node list (`colab_pf_data/pf_balance_nodes_refined.csv`), label controls, `flow_relative` default.

### Fine-tune knobs when init eval is already strong (|V| MAE < 0.004)

If baseline eval shows |V| MAE below ~0.004, treat physics as a gentle nudge — not a full retrain:

| Knob | Recommended | Why |
|---|---|---|
| `FINETUNE_PHYSICS_WEIGHT` | **0.01–0.02** | Full 0.05 at epoch 1 can dominate the weak voltage anchor and spike |V| MAE. |
| `FINETUNE_LAMBDA_VOLTAGE` | **0.05–0.1** | Keeps the pretrained voltage head anchored while physics ramps in. |
| `PF_WEIGHT_WARMUP_EPOCHS` | **3** | Zero physics for the first epochs so the loaded checkpoint is not immediately pulled off-manifold. |
| `PF_WEIGHT_RAMP_EPOCHS` | **5** | Linear ramp to full physics weight after warmup. |
| `FINETUNE_LR` | **5e-5** (default) | Lower (1e-5) if epoch-1 `[vs baseline]` still regresses. |

Trainer loads **norm stats** (`x/y/reg/pv_mean.pt`, `*_std.pt`) from `INIT_RUN_DIR` automatically when present (see upload list above). Ensure those sidecars are uploaded with the baseline run — recomputing stats from a different split/seed misaligns the checkpoint's normalized head.


````python
# --- Section 8: Engage-style physics fine-tune from saved baseline ---
import datetime
import json
import os
import subprocess
import sys
import warnings
from pathlib import Path

DRIVE_ROOT = Path("/content/drive")
MYDRIVE_DATA = DRIVE_ROOT / "MyDrive/datasets_gnn2"
COLAB_CHUNK_DEFAULT = MYDRIVE_DATA / "original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
WIN_CHUNK_DEFAULT = Path(r"datasets_gnn2_from pc/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")


def _on_colab() -> bool:
    return os.environ.get("COLAB_RELEASE_TAG") is not None


def _drive_mounted() -> bool:
    return DRIVE_ROOT.is_dir() and (DRIVE_ROOT / "MyDrive").is_dir()


def _find_repo_root() -> Path:
    env = os.environ.get("GNN2_REPO_ROOT", "").strip()
    if env:
        p = Path(env).expanduser().resolve()
        if (p / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
            return p
    for cand in (Path.cwd(), Path("/content/GNN-Sandia")):
        cand = cand.resolve()
        if (cand / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
            return cand
    raise FileNotFoundError("Could not find repo root (train_da_gps_multitask_complex_voltage_gine.py)")


def _resolve_data_path(path: Path, *, label: str, colab_fallback: Path | None = None) -> Path:
    p = Path(path).expanduser()
    if not p.is_absolute():
        p = (REPO / p).resolve()
    else:
        p = p.resolve()
    if p.exists():
        return p
    if colab_fallback is not None and Path(colab_fallback).exists():
        warnings.warn(f"{label}: {p} missing; using Colab fallback {colab_fallback}", stacklevel=2)
        return Path(colab_fallback).resolve()
    return p


def _find_node_pe_csv(chunk_parent: Path) -> Path:
    hits = sorted(chunk_parent.glob("run_*/gnn_node_index_master.csv"))
    if not hits:
        raise FileNotFoundError(f"No run_*/gnn_node_index_master.csv under {chunk_parent}")
    return hits[0]


def _chunks_from_subdir_glob(chunk_parent: Path, glob_pat: str) -> list[Path]:
    import fnmatch

    glob_pat = str(glob_pat).strip()
    if "," in glob_pat:
        allowed = {s.strip() for s in glob_pat.split(",") if s.strip()}
        return sorted(
            (p for p in chunk_parent.iterdir() if p.is_dir() and p.name in allowed),
            key=lambda p: p.name,
        )
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and fnmatch.fnmatch(p.name, glob_pat)),
        key=lambda p: p.name,
    )


def _run_trainer(cmd: list[str]) -> None:
    print("\nRunning:\n ", " ".join(cmd), "\n", flush=True)
    with subprocess.Popen(
        cmd,
        cwd=str(REPO),
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        bufsize=1,
        text=True,
    ) as proc:
        assert proc.stdout is not None
        for line in proc.stdout:
            print(line, end="")
        rc = proc.wait()
    if rc != 0:
        raise subprocess.CalledProcessError(rc, cmd)


_INIT_CHECKPOINT_NAMES = ("da_gps_multitask_best.pt", "training_last.pt")
_INIT_CKPT_STATS_NAMES = frozenset(
    {
        "x_mean.pt",
        "x_std.pt",
        "y_mean.pt",
        "y_std.pt",
        "reg_mean.pt",
        "reg_std.pt",
        "pv_mean.pt",
        "pv_std.pt",
        "reg_class_values.pt",
    }
)


def _abs_repo_path(p: Path, repo: Path) -> Path:
    p = Path(p).expanduser()
    if not p.is_absolute():
        p = (repo / p).resolve()
    else:
        p = p.resolve()
    return p


def _print_init_run_dir_listing(label: str, p: Path, repo: Path) -> None:
    """Pre-flight: list files in a candidate baseline run directory."""
    p = _abs_repo_path(p, repo)
    print(f"\n--- {label}: {p} ---")
    if not p.is_dir():
        print("  (directory not found)")
        return
    entries = sorted(p.iterdir(), key=lambda x: (not x.is_file(), x.name.lower()))
    if not entries:
        print("  (empty directory)")
        return
    for item in entries:
        if item.is_dir():
            print(f"  {item.name}/")
        else:
            mb = item.stat().st_size / (1024 * 1024)
            print(f"  {item.name}  ({mb:.2f} MB)")


def _scan_run_dir_model_checkpoints(run_dir: Path, *, tried: list[str]) -> Path | None:
    """Last resort: newest non-stats .pt in run_dir (mid-train snapshots)."""
    if not run_dir.is_dir():
        return None
    pts = [
        p
        for p in run_dir.glob("*.pt")
        if p.name not in _INIT_CHECKPOINT_NAMES and p.name not in _INIT_CKPT_STATS_NAMES
    ]
    pts.sort(key=lambda p: p.stat().st_mtime, reverse=True)
    for p in pts:
        tried.append(f"INIT_CHECKPOINT (scan): {p}")
        if p.is_file() and p.stat().st_size > 0:
            return p
    return None



def _drive_search_init_artifacts(run_name: str) -> dict[str, list[Path]]:
    """Walk MYDRIVE_DATA for run folders or checkpoint files matching run_name."""
    out: dict[str, list[Path]] = {"run_dirs": [], "checkpoints": []}
    if not (_on_colab() and _drive_mounted()):
        return out
    root = MYDRIVE_DATA
    if not root.is_dir():
        return out
    seen_dirs: set[str] = set()
    seen_ckpt: set[str] = set()
    try:
        for dirpath, dirnames, filenames in os.walk(root):
            base = Path(dirpath).name
            if base == run_name and Path(dirpath).is_dir():
                key = str(Path(dirpath).resolve())
                if key not in seen_dirs:
                    seen_dirs.add(key)
                    out["run_dirs"].append(Path(dirpath).resolve())
            for fn in filenames:
                if fn not in _INIT_CHECKPOINT_NAMES:
                    continue
                p = Path(dirpath) / fn
                if run_name in str(p) or base == run_name:
                    key = str(p.resolve())
                    if key not in seen_ckpt:
                        seen_ckpt.add(key)
                        out["checkpoints"].append(p.resolve())
    except OSError as exc:
        warnings.warn(f"Drive search under {root} failed: {exc}", stacklevel=2)
    return out


def _print_drive_search_results(run_name: str, search: dict[str, list[Path]]) -> None:
    """Print Drive search hits before failing path resolution."""
    run_dirs = search.get("run_dirs") or []
    ckpts = search.get("checkpoints") or []
    if not run_dirs and not ckpts:
        print(f"\n=== Drive search: no folders or checkpoints matching {run_name!r} under {MYDRIVE_DATA} ===")
        return
    print(f"\n=== Drive search under {MYDRIVE_DATA} (run_name={run_name!r}) ===")
    if run_dirs:
        print("  Matching run directories:")
        for p in run_dirs:
            print(f"    {p}")
    else:
        print("  Matching run directories: (none)")
    if ckpts:
        print("  Matching checkpoint files:")
        for p in ckpts:
            print(f"    {p}")
    else:
        print("  Matching checkpoint files: (none)")

def _resolve_init_paths(
    *,
    run_dir: Path,
    checkpoint: Path,
    repo: Path,
    runs_parent: Path | None = None,
) -> tuple[Path, Path]:
    """Resolve INIT_RUN_DIR and INIT_CHECKPOINT with env overrides and Drive fallbacks."""
    run_name = run_dir.name or _INIT_RUN_BASENAME
    tried: list[str] = []

    env_run = os.environ.get("GNN2_INIT_RUN_DIR", "").strip()
    env_ckpt = os.environ.get("GNN2_INIT_CHECKPOINT", "").strip()

    def _abs(p: Path) -> Path:
        return _abs_repo_path(p, repo)

    run_candidates: list[Path] = []
    if env_run:
        run_candidates.append(Path(env_run))
    run_candidates.append(run_dir)
    if runs_parent is not None:
        run_candidates.append(runs_parent / run_name)
    if _on_colab() and _drive_mounted():
        run_candidates.extend(
            [
                MYDRIVE_DATA / "runs" / run_name,
                MYDRIVE_DATA / "attention checkpoints" / run_name,
            ]
        )
    run_candidates.extend(
        [
            repo / "gnn2_architecture_search/attention checkpoints" / run_name,
            Path("/content/GNN-Sandia/gnn2_architecture_search/attention checkpoints") / run_name,
        ]
    )

    seen_run: set[str] = set()
    unique_runs: list[Path] = []
    for cand in run_candidates:
        p = _abs(cand)
        key = str(p)
        if key in seen_run:
            continue
        seen_run.add(key)
        unique_runs.append(p)

    resolved_run: Path | None = None
    existing_runs: list[Path] = []
    for p in unique_runs:
        tried.append(f"INIT_RUN_DIR: {p}")
        if p.is_dir():
            existing_runs.append(p)
            if resolved_run is None:
                resolved_run = p

    ckpt_candidates: list[Path] = []
    if env_ckpt:
        ckpt_candidates.append(Path(env_ckpt))
    ckpt_candidates.append(checkpoint)
    for parent in unique_runs:
        for name in _INIT_CHECKPOINT_NAMES:
            ckpt_candidates.append(parent / name)

    resolved_ckpt: Path | None = None
    seen_ckpt: set[str] = set()
    for cand in ckpt_candidates:
        p = _abs(cand)
        key = str(p)
        if key in seen_ckpt:
            continue
        seen_ckpt.add(key)
        tried.append(f"INIT_CHECKPOINT: {p}")
        if p.is_file():
            resolved_ckpt = p
            break

    if resolved_ckpt is None:
        for parent in existing_runs or unique_runs:
            hit = _scan_run_dir_model_checkpoints(parent, tried=tried)
            if hit is not None:
                resolved_ckpt = hit
                if resolved_run is None:
                    resolved_run = parent
                break

    if resolved_run is None and resolved_ckpt is not None:
        resolved_run = resolved_ckpt.parent
        tried.append(f"INIT_RUN_DIR (from checkpoint parent): {resolved_run}")

    if resolved_run is None or resolved_ckpt is None:
        missing = []
        if resolved_run is None:
            missing.append("INIT_RUN_DIR")
        if resolved_ckpt is None:
            missing.append("INIT_CHECKPOINT")

        drive_search = _drive_search_init_artifacts(run_name)
        _print_drive_search_results(run_name, drive_search)

        # Retry candidates discovered on Drive before failing.
        for p in drive_search.get("run_dirs") or []:
            if p.is_dir() and resolved_run is None:
                resolved_run = p
                tried.append(f"INIT_RUN_DIR (Drive search): {p}")
        for p in drive_search.get("checkpoints") or []:
            if p.is_file() and resolved_ckpt is None:
                resolved_ckpt = p
                tried.append(f"INIT_CHECKPOINT (Drive search): {p}")
                if resolved_run is None:
                    resolved_run = p.parent
        if resolved_run is not None and resolved_ckpt is not None:
            return resolved_run, resolved_ckpt

        drive_hint = (
            f"{MYDRIVE_DATA / 'runs' / run_name}"
            if _on_colab()
            else f"<MYDRIVE_DATA>/runs/{run_name}"
        )
        parent_ok = runs_parent is not None and runs_parent.is_dir()
        if _on_colab() and parent_ok and not drive_search["run_dirs"] and not drive_search["checkpoints"]:
            cause = (
                "No baseline checkpoint on Google Drive for this run name. "
                "The Colab VM is ephemeral — if training wrote only under "
                "/content/GNN-Sandia, that copy is gone after disconnect. "
                "Copy the run folder to Drive or retrain Section 3."
            )
        else:
            cause = (
                "Configured init paths were not found at any standard location. "
                "Check RUNS_PARENT / INIT_RUN_DIR below, or set GNN2_INIT_RUN_DIR / "
                "GNN2_INIT_CHECKPOINT."
            )
        raise FileNotFoundError(
            f"{' and '.join(missing)} not found.\n"
            f"Likely cause: {cause}\n\n"
            f"Paths tried:\n  " + "\n  ".join(tried)
            + f"\n\nExpected Drive location:\n  {drive_hint}\n"
            + (f"RUNS_PARENT exists: {runs_parent}\n" if runs_parent else "")
            + "Set GNN2_INIT_RUN_DIR / GNN2_INIT_CHECKPOINT to override."
        )

    if resolved_run != _abs(run_dir):
        warnings.warn(
            f"Using INIT_RUN_DIR {resolved_run} (configured {run_dir})",
            stacklevel=2,
        )
    if resolved_ckpt != _abs(checkpoint):
        warnings.warn(
            f"Using INIT_CHECKPOINT {resolved_ckpt} (configured {checkpoint})",
            stacklevel=2,
        )
    return resolved_run, resolved_ckpt


from train_da_gps_multitask_complex_voltage_gine import (
    _baseline_report_specs_for,
    _format_baseline_metric_value,
    _metric_delta_status,
    _print_eval_metrics_line,
)


def _print_metrics_table(report: dict, *, title: str) -> None:
    print(f"\n=== {title} ===")
    for split in ("val", "test"):
        met = report.get(f"{split}_metrics") or report.get(f"init_baseline_{split}_metrics")
        if not met:
            continue
        _print_eval_metrics_line(split.capitalize(), met)


# --- configurable ---
_INIT_RUN_BASENAME = "da_gps_chunked_l4_mvagg_gine_metaaux_regce_20260516_225149_CCE"
INIT_RUN_DIR = Path(
    "gnn2_architecture_search/attention checkpoints/" + _INIT_RUN_BASENAME
)
INIT_CHECKPOINT = INIT_RUN_DIR / "da_gps_multitask_best.pt"
FINETUNE_PF_LOSS_MODE = "flow_relative"  # or hybrid, absolute
FINETUNE_PHYSICS_WEIGHT = 0.02  # 0.01-0.02 when init |V| MAE < 0.004
FINETUNE_PF_ABSOLUTE_WEIGHT = 0.1  # hybrid alpha
FINETUNE_EPOCHS = 20
FINETUNE_LR = 5e-5
FINETUNE_LAMBDA_VOLTAGE = 0.05  # voltage anchor; raise to 0.1 if physics still dominates
PF_WEIGHT_WARMUP_EPOCHS = 3
PF_WEIGHT_RAMP_EPOCHS = 5
CHUNK_GLOB = "run_*"  # full 40 chunks
SMOKE_TEST = False
SAVE_ONLY_IF_IMPROVED = True

# Must match init checkpoint (local shipped baseline: h=96, node_emb_dim=4; legacy folder name has "l4")
LIGHTER_HIDDEN = 96
LIGHTER_LAYERS = 2
LIGHTER_HEADS = 2
LIGHTER_NODE_EMB_DIM = 4
META_AUX_COLS = "pv_pv2_p_post_kw,pv_pv2_q_post_kvar,P_loss_total_post_kw,Q_loss_total_post_kvar"
PF_BALANCE_NODE_LIST_REL = "colab_pf_data/pf_balance_nodes_refined.csv"
NUM_WORKERS = 0 if os.name == "nt" else 4

if _on_colab() and not _drive_mounted():
    from google.colab import drive

    drive.mount("/content/drive")

REPO = _find_repo_root()
os.chdir(REPO)
sys.path.insert(0, str(REPO))
os.environ["PYTHONUNBUFFERED"] = "1"
os.environ["GNN2_REPO_ROOT"] = str(REPO)

if _on_colab():
    if not _drive_mounted():
        raise RuntimeError(
            "Colab requires Google Drive mounted. Re-run this cell and approve Drive access, "
            "or: from google.colab import drive; drive.mount('/content/drive')"
        )
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    DA_CACHE_ROOT = MYDRIVE_DATA / "cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = MYDRIVE_DATA / "runs"  # persisted on Drive; /content/GNN-Sandia is ephemeral
elif os.name == "nt":
    CHUNK_PARENT = WIN_CHUNK_DEFAULT
    DA_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
else:
    CHUNK_PARENT = REPO / "datasets_gnn2_from pc/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
    DA_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/da_gps_chunked_mvagg_full_gine"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"

_colab_fb = COLAB_CHUNK_DEFAULT if _on_colab() else None
chunk_parent = _resolve_data_path(CHUNK_PARENT, label="CHUNK_PARENT", colab_fallback=_colab_fb)
da_cache_root = _resolve_data_path(DA_CACHE_ROOT, label="DA_CACHE_ROOT", colab_fallback=MYDRIVE_DATA / "cache/da_gps_chunked_mvagg_full_gine" if _on_colab() else None)
gnn_cache_root = _resolve_data_path(GNN_CACHE_ROOT, label="GNN_CACHE_ROOT", colab_fallback=MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine" if _on_colab() else None)
runs_parent = _resolve_data_path(RUNS_PARENT, label="RUNS_PARENT", colab_fallback=MYDRIVE_DATA / "runs" if _on_colab() else None)

print("=== Section 8 INIT_RUN_DIR preflight (before resolve) ===")
print(f"RUNS_PARENT (parent only): {runs_parent}")
print(f"INIT run basename: {_INIT_RUN_BASENAME}")
for _lbl, _cand in (
    ("configured INIT_RUN_DIR", INIT_RUN_DIR),
    (f"RUNS_PARENT / {_INIT_RUN_BASENAME}", runs_parent / _INIT_RUN_BASENAME),
    ("repo attention checkpoints", REPO / "gnn2_architecture_search/attention checkpoints" / _INIT_RUN_BASENAME),
    ("Colab clone (ephemeral)", Path("/content/GNN-Sandia/gnn2_architecture_search/attention checkpoints") / _INIT_RUN_BASENAME),
    ("Drive runs", MYDRIVE_DATA / "runs" / _INIT_RUN_BASENAME),
    ("Drive attention checkpoints", MYDRIVE_DATA / "attention checkpoints" / _INIT_RUN_BASENAME),
):
    _print_init_run_dir_listing(_lbl, _cand, REPO)

init_run_dir, init_checkpoint = _resolve_init_paths(
    run_dir=INIT_RUN_DIR,
    checkpoint=INIT_CHECKPOINT,
    repo=REPO,
    runs_parent=runs_parent,
)

if SMOKE_TEST:
    runs = sorted(p.name for p in chunk_parent.iterdir() if p.name.startswith("run_"))
    CHUNK_GLOB = ",".join(runs[:3])
    FINETUNE_EPOCHS = 5

chunks = _chunks_from_subdir_glob(chunk_parent, CHUNK_GLOB)
node_pe = _find_node_pe_csv(chunk_parent)

from gnn2_pf_data_paths import resolve_pf_catalog_paths

_reg_cat, _cap_map, PF_DATA_ROOT = resolve_pf_catalog_paths(
    repo=REPO,
    preferred_root=None,
    chunk_parent=chunk_parent,
)

_pf_bal = (REPO / PF_BALANCE_NODE_LIST_REL).resolve()
if not _pf_bal.is_file():
    raise FileNotFoundError(f"PF balance list missing: {_pf_bal}")

tag = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
finetune_out = runs_parent / (
    f"da_gps_finetune_pf_{FINETUNE_PF_LOSS_MODE}_w{FINETUNE_PHYSICS_WEIGHT:g}_from_baseline_{tag}"
)
finetune_out.mkdir(parents=True, exist_ok=True)

_base_cmd = [
    sys.executable,
    "-u",
    "train_da_gps_multitask_complex_voltage_gine.py",
    "--chunk_parent",
    str(chunk_parent),
    "--chunk_subdir_glob",
    CHUNK_GLOB,
    "--nodes_csv",
    "gnn_node_features_and_targets_mvagg.csv",
    "--edge_catalog_csv",
    "gnn_edges_phase_static.csv",
    "--meta_csv",
    "gnn_sample_meta.csv",
    "--node_feature_cols",
    "p_load_kw,q_load_kvar,p_pv_kw",
    "--exclude_bess_features",
    "--node_pe_csv",
    str(node_pe),
    "--node_pe_cols",
    "auto",
    "--n_system_tokens",
    "10",
    "--aux_meta_cols",
    META_AUX_COLS,
    "--out_dir",
    str(finetune_out),
    "--cache_dir",
    str(da_cache_root),
    "--bootstrap_gnn_cache_dir",
    str(gnn_cache_root),
    "--hidden",
    str(LIGHTER_HIDDEN),
    "--layers",
    str(LIGHTER_LAYERS),
    "--heads",
    str(LIGHTER_HEADS),
    "--node_emb_dim",
    str(LIGHTER_NODE_EMB_DIM),
    "--edge_emb_dim",
    "0",
    "--reg_loss",
    "ce",
    "--per_device_cap_head",
    "--per_device_reg_head",
    "--train_frac",
    "0.80",
    "--val_frac",
    "0.10",
    "--sample_frac",
    "1.0",
    "--num_workers",
    str(NUM_WORKERS),
    "--seed",
    "42",
    "--init_checkpoint",
    str(init_checkpoint),
    "--init_run_dir",
    str(init_run_dir),
]

print("=== Section 8 preflight ===")
print(f"REPO:            {REPO}")
print(f"INIT_RUN_DIR:    {init_run_dir}")
print(f"INIT_CHECKPOINT: {init_checkpoint}")
print(f"CHUNK_PARENT:    {chunk_parent}  ({len(chunks)} chunks)")
print(f"DA cache:        {da_cache_root}")
print(f"PF balance:      {_pf_bal}")
print(f"Fine-tune out:   {finetune_out}")

# Step 1: baseline metrics without training
eval_cmd = _base_cmd + [
    "--eval_only",
    "--epochs",
    "0",
    "--batch_size",
    "64",
]
_run_trainer(eval_cmd)
baseline_report = json.loads((finetune_out / "da_gps_report.json").read_text(encoding="utf-8"))
_print_metrics_table(baseline_report, title="Baseline before post-tuning (init checkpoint)")

# Step 2: physics fine-tune
ft_cmd = _base_cmd + [
    "--epochs",
    str(FINETUNE_EPOCHS),
    "--batch_size",
    "64",
    "--lr",
    str(FINETUNE_LR),
    "--weight_decay",
    "1e-5",
    "--lambda_voltage",
    str(FINETUNE_LAMBDA_VOLTAGE),
    "--finetune_physics_only",
    "--lambda_cap",
    "0",
    "--lambda_reg",
    "0",
    "--lambda_pv",
    "0",
    "--patience",
    str(max(10, FINETUNE_EPOCHS)),
    "--early_stop_on",
    "voltage",
    "--log_every",
    "1",
    "--pf_data_root",
    str(PF_DATA_ROOT),
    "--loss_power_balance_weight",
    str(FINETUNE_PHYSICS_WEIGHT),
    "--pf_loss_mode",
    FINETUNE_PF_LOSS_MODE,
    "--pf_balance_node_list_csv",
    str(_pf_bal),
    "--pf_sparse_y",
    "1",
    "--pf_huber_delta_kw",
    "10",
    "--pf_detach_controls",
    "--pf_weight_warmup_epochs",
    str(PF_WEIGHT_WARMUP_EPOCHS),
    "--pf_weight_ramp_epochs",
    str(PF_WEIGHT_RAMP_EPOCHS),
]
if FINETUNE_PF_LOSS_MODE == "hybrid":
    ft_cmd.extend(["--pf_absolute_weight", str(FINETUNE_PF_ABSOLUTE_WEIGHT)])
if SAVE_ONLY_IF_IMPROVED:
    ft_cmd.append("--save_only_if_improved")

_run_trainer(ft_cmd)
ft_report = json.loads((finetune_out / "da_gps_report.json").read_text(encoding="utf-8"))

print("\n=== Fine-tune result ===")
print(f"checkpoint_saved: {ft_report.get('checkpoint_saved')}")
if ft_report.get("checkpoint_reject_reason"):
    print(f"REJECTED: {ft_report['checkpoint_reject_reason']}")
_print_metrics_table(ft_report, title="After fine-tune (report)")

if ft_report.get("init_baseline_test_metrics") and ft_report.get("test_metrics"):
    print("\n=== Before / after (from da_gps_report.json) ===")
    print(
        f"{'split':<6} {'metric':<22} {'baseline':>12} {'after':>12} {'delta':>12} {'status':>10}"
    )
    for split, bkey, akey in (
        ("val", "init_baseline_val_metrics", "val_metrics"),
        ("test", "init_baseline_test_metrics", "test_metrics"),
    ):
        bmet = ft_report.get(bkey) or {}
        amet = ft_report.get(akey) or {}
        for key, label, higher_is_better in _baseline_report_specs_for(bmet):
            b = float(bmet.get(key, float("nan")))
            a = float(amet.get(key, float("nan")))
            if b != b and a != a:
                continue
            delta = a - b
            status = _metric_delta_status(delta, higher_is_better=higher_is_better)
            delta_s = f"{delta:+.4f}" if key.startswith(("loss_", "r2_")) else f"{delta:+.6f}"
            print(
                f"{split:<6} {label:<22} "
                f"{_format_baseline_metric_value(key, b):>12} "
                f"{_format_baseline_metric_value(key, a):>12} "
                f"{delta_s:>12} {status:>10}"
            )

print("\nDone.")
print("Fine-tune out_dir:", finetune_out.resolve())
````

## Full DA-GPS training with add-ons (section 3 extension)

**This cell runs full production training** — not a smoke test. Defaults: all **40** `run_*` chunks, **200** epochs, patience **30**, `sample_frac=1.0`, seed **42**, cache `da_gps_chunked_mvagg_full_gine`.

Three training add-ons are **enabled** (locked-in architectural decisions). Base section-3 cell leaves them off; this extension turns all three on atop the lighter GINE stack.

### Architecture (locked in)

| Setting | Value |
|---------|-------|
| `hidden` / `layers` / `heads` | 64 / 2 / 2 |
| `node_emb_dim` / `edge_emb_dim` | 2 / 0 |
| `PHYSICS_WEIGHT` | `0.0` (baseline; `early_stop_on=total`) |
| `reg_loss` | `ce` + `--per_device_reg_head` |

### Add-ons (all on)

| Add-on | Flags | Notes |
|--------|-------|-------|
| Regulator territory attention bias | `--attn_reg_territory_bias`, `--attn_reg_territory_beta 2.0` | 2nd cross-attn only; regulator token **keys** |
| Ordinal / cost-sensitive regulator CE | `--reg_ordinal_ce` | requires `--reg_loss ce` |
| Violation-weighted voltage MSE | `--volt_violation_weight`, `--volt_lo_pu 0.96`, `--volt_hi_pu 1.04`, `--volt_violation_alpha 2.0` | up-weights nodes with target \|V\| outside band |

**Territory bias** uses hop-downstream distance from repo `datasets_gnn2_from pc/load_hop_distance_to_each_regulator_all_index_nodes.csv` (or `--hop_csv`).

**Ordinal CE** uses cost-sensitive CE with cost = \|tap_j − tap_c\| from `reg_class_tables`.

**Violation weights** up-weight voltage MSE on nodes whose **target** \|V\| is outside [lo, hi] pu.

### Smoke vs full (toggle at top of code cell)

| `SMOKE_TEST` | Chunks | `epochs` | `patience` | Cache | Run-folder suffix |
|---|---|---|---|---|---|
| `False` (**default here**) | all `run_*` (40) | 200 | 30 | `da_gps_chunked_mvagg_full_gine` | `_addons_full` |
| `True` | first `SMOKE_CHUNK_COUNT` `run_*` | `SMOKE_EPOCHS` | `SMOKE_PATIENCE` | `da_gps_chunked_mvagg_smoke_gine` | `_addons_smoke` |

### Paths (Colab primary)

| | Colab (Drive + GNN-Sandia clone) | Local Windows / Linux |
|---|---|---|
| `CHUNK_PARENT` | `/content/drive/MyDrive/datasets_gnn2/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40` | auto-detect (see code cell) |
| `NODE_PE_CSV` | `.../run_001_scen_0000_0049_seed_20420233/gnn_node_index_master.csv` | auto: first `run_*/...` |
| Caches | `/content/drive/MyDrive/datasets_gnn2/cache/da_gps_chunked_mvagg_full_gine` (persistent on Drive) | repo `datasets_gnn2_from pc/cache/` |
| Checkpoints | `/content/GNN-Sandia/gnn2_architecture_search/attention checkpoints/` | repo `gnn2_architecture_search/attention checkpoints/` |

Run folder name pattern: `da_gps_chunked_l2_h64_mvagg_gine_metaaux_regce_addons_full_{timestamp}`. Batch/epoch logs and `da_gps_report.json` → `training_addons` when any add-on is on.

**Note:** `/content/GNN-Sandia/...` checkpoints are ephemeral on Colab disconnect; copy to MyDrive if you need them for Section 8.

````python
# --- DA-GPS full training with add-ons enabled (NOT smoke) ---
import os
import sys
import subprocess
import datetime
import warnings
from pathlib import Path

DRIVE_ROOT = Path("/content/drive")
MYDRIVE_DATA = DRIVE_ROOT / "MyDrive/datasets_gnn2"
COLAB_CHUNK_DEFAULT = MYDRIVE_DATA / "original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
COLAB_NODE_PE_DEFAULT = (
    COLAB_CHUNK_DEFAULT
    / "run_001_scen_0000_0049_seed_20420233/gnn_node_index_master.csv"
)
COLAB_RUNS_PARENT = Path("/content/GNN-Sandia/gnn2_architecture_search/attention checkpoints")
WIN_CHUNK_DEFAULT = Path(r"D:\datasets\original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")


def _is_windows_drive_path(path) -> bool:
    s = str(path).strip().replace("/", "\\")
    return len(s) >= 2 and s[1] == ":" and s[0].isalpha()


def _on_colab() -> bool:
    return Path("/content").is_dir()


def _drive_mounted() -> bool:
    return DRIVE_ROOT.is_dir() and (DRIVE_ROOT / "MyDrive").is_dir()


def _resolve_data_path(path: Path, *, label: str, colab_fallback: Path | None = None) -> Path:
    """Resolve a data path; never join Windows drive letters with cwd on Linux."""
    raw = str(path)
    if _is_windows_drive_path(raw):
        if _on_colab() and colab_fallback is not None:
            warnings.warn(
                f"{label}={raw!r} is a Windows path on Linux/Colab; using {colab_fallback} instead.",
                UserWarning,
                stacklevel=2,
            )
            return colab_fallback.expanduser().resolve()
        if _on_colab():
            raise ValueError(
                f"{label}={raw!r} is a Windows absolute path and invalid on Colab. "
                f"Mount Drive and use e.g. {COLAB_CHUNK_DEFAULT}"
            )
        return Path(raw).expanduser().resolve()
    p = Path(path).expanduser()
    if p.is_absolute():
        return p.resolve()
    return (Path.cwd() / p).resolve()


def _find_node_pe_csv(chunk_parent: Path) -> Path:
    hits = sorted(chunk_parent.glob("run_*/gnn_node_index_master.csv"))
    if not hits:
        raise FileNotFoundError(
            f"No run_*/gnn_node_index_master.csv under {chunk_parent}"
        )
    return hits[0]

def _sorted_run_chunk_dirs(chunk_parent: Path) -> list[Path]:
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_")),
        key=lambda p: p.name,
    )


def _smoke_chunk_subdir_glob(chunk_parent: Path, count: int) -> str:
    """First `count` run_* names, comma-separated for --chunk_subdir_glob."""
    names = [p.name for p in _sorted_run_chunk_dirs(chunk_parent)]
    if len(names) < count:
        raise ValueError(
            f"SMOKE_CHUNK_COUNT={count} but only {len(names)} run_* under {chunk_parent}"
        )
    return ",".join(names[:count])


def _chunks_from_subdir_glob(chunk_parent: Path, glob_pat: str) -> list[Path]:
    import fnmatch

    glob_pat = str(glob_pat).strip()
    if "," in glob_pat:
        allowed = {s.strip() for s in glob_pat.split(",") if s.strip()}
        chunks = sorted(
            (p for p in chunk_parent.iterdir() if p.is_dir() and p.name in allowed),
            key=lambda p: p.name,
        )
        missing = allowed - {p.name for p in chunks}
        if missing:
            raise FileNotFoundError(f"Missing smoke chunk folders: {sorted(missing)}")
        return chunks
    return sorted(
        (p for p in chunk_parent.iterdir() if p.is_dir() and fnmatch.fnmatch(p.name, glob_pat)),
        key=lambda p: p.name,
    )


def _write_smoke_compare_manifest(
    runs_parent: Path,
    *,
    out_dir: Path,
    physics_weight: float,
    chunk_glob: str,
    n_chunks: int,
    epochs: int,
    seed: int,
) -> None:
    import json as _json

    path = runs_parent / "_smoke_compare_manifest.json"
    manifest = {}
    if path.is_file():
        manifest = _json.loads(path.read_text(encoding="utf-8"))
    key = "physics" if physics_weight > 0 else "baseline"
    manifest[key] = {
        "run_dir": str(out_dir.resolve()),
        "physics_weight": physics_weight,
        "chunk_glob": chunk_glob,
        "n_chunks": n_chunks,
        "epochs": epochs,
        "seed": seed,
    }
    path.write_text(_json.dumps(manifest, indent=2), encoding="utf-8")
    print(f"Smoke compare manifest ({key}) -> {path}")


# Colab: mount Drive when chunks/caches live under MyDrive
if _on_colab() and not _drive_mounted():
    from google.colab import drive
    drive.mount("/content/drive")

# --- FULL training (set True only for quick smoke/debug) ---
SMOKE_TEST = False
SMOKE_CHUNK_COUNT = 3  # used only when SMOKE_TEST=True
SMOKE_EPOCHS = 15
SMOKE_PATIENCE = 5
SMOKE_SEED = 42
FULL_EPOCHS = 200
FULL_PATIENCE = 30
TRAIN_SEED = 42
_DA_CACHE_NAME = "da_gps_chunked_mvagg_smoke_gine" if SMOKE_TEST else "da_gps_chunked_mvagg_full_gine"

# --- paths: auto-detect Colab + Drive; override below if needed ---
_env_repo = os.environ.get("GNN2_REPO_ROOT", "").strip()
REPO = Path(_env_repo).expanduser().resolve() if _env_repo else Path.cwd().resolve()
if not (REPO / "train_da_gps_multitask_complex_voltage_gine.py").is_file():
    raise FileNotFoundError(f"Repo root missing trainer script: {REPO}")

if _on_colab():
    if not _drive_mounted():
        raise RuntimeError(
            "Colab requires Google Drive mounted. Re-run this cell and approve Drive access, "
            "or: from google.colab import drive; drive.mount('/content/drive')"
        )
    CHUNK_PARENT = COLAB_CHUNK_DEFAULT
    NODE_PE_CSV = COLAB_NODE_PE_DEFAULT
    # Caches must live on Drive to reuse across Colab sessions (/content is wiped on reset)
    DA_CACHE_ROOT = MYDRIVE_DATA / f"cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = COLAB_RUNS_PARENT
elif os.name == "nt":
    CHUNK_PARENT = WIN_CHUNK_DEFAULT
    NODE_PE_CSV = None  # auto: first run_*/gnn_node_index_master.csv under CHUNK_PARENT
    DA_CACHE_ROOT = REPO / f"datasets_gnn2_from pc/cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"
else:
    CHUNK_PARENT = REPO / "datasets_gnn2_from pc/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40"
    NODE_PE_CSV = None
    DA_CACHE_ROOT = REPO / f"datasets_gnn2_from pc/cache/{_DA_CACHE_NAME}"
    GNN_CACHE_ROOT = REPO / "datasets_gnn2_from pc/cache/gnn_only_chunked_mvagg_full_gine"
    RUNS_PARENT = REPO / "gnn2_architecture_search/attention checkpoints"

# Optional overrides (use POSIX paths on Colab, not D:\...):
# CHUNK_PARENT = Path("/content/drive/MyDrive/datasets_gnn2/original_8500_unbalanced_chunked_no_bess_new_diverse_2000_40")
# NODE_PE_CSV = CHUNK_PARENT / "run_001_scen_0000_0049_seed_20420233/gnn_node_index_master.csv"
# RUNS_PARENT = MYDRIVE_DATA / "runs"  # persist checkpoints on Drive instead of ephemeral clone

# --- lighter architecture (locked in for add-ons full run) ---
LIGHTER_HIDDEN = 64
LIGHTER_LAYERS = 2
LIGHTER_HEADS = 2
LIGHTER_NODE_EMB_DIM = 2
LIGHTER_EDGE_EMB_DIM = 0

PHYSICS_WEIGHT = 0.0  # baseline; early_stop_on=total

# Explicit PF balance nodes when physics on (1177 hetero MV load nodes; chunk-safe)
PF_BALANCE_NODE_LIST_CSV = REPO / "colab_pf_data/pf_balance_nodes_explicit.csv"

META_AUX_COLS = "pv_pv2_p_post_kw,pv_pv2_q_post_kvar,P_loss_total_post_kw,Q_loss_total_post_kvar"
NUM_WORKERS = 0 if os.name == "nt" else 4

os.chdir(REPO)
sys.path.insert(0, str(REPO))
os.environ["PYTHONUNBUFFERED"] = "1"
os.environ.setdefault("GNN2_REPO_ROOT", str(REPO))

_colab_fb = COLAB_CHUNK_DEFAULT if _on_colab() else None
chunk_parent = _resolve_data_path(CHUNK_PARENT, label="CHUNK_PARENT", colab_fallback=_colab_fb)
if SMOKE_TEST:
    CHUNK_GLOB = _smoke_chunk_subdir_glob(chunk_parent, SMOKE_CHUNK_COUNT)
    EPOCHS = SMOKE_EPOCHS
    PATIENCE = SMOKE_PATIENCE
    SEED = SMOKE_SEED
else:
    CHUNK_GLOB = "run_*"
    EPOCHS = FULL_EPOCHS
    PATIENCE = FULL_PATIENCE
    SEED = TRAIN_SEED
da_cache_root = _resolve_data_path(
    DA_CACHE_ROOT,
    label="DA_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / f"cache/{_DA_CACHE_NAME}" if _on_colab() else None,
)
gnn_cache_root = _resolve_data_path(
    GNN_CACHE_ROOT,
    label="GNN_CACHE_ROOT",
    colab_fallback=MYDRIVE_DATA / "cache/gnn_only_chunked_mvagg_full_gine" if _on_colab() else None,
)
runs_parent = _resolve_data_path(
    RUNS_PARENT,
    label="RUNS_PARENT",
    colab_fallback=MYDRIVE_DATA / "runs" if _on_colab() else None,
)

# --- PF topology root (repo colab_pf_data/ after git pull, or Drive dailyagg fallback) ---
from gnn2_pf_data_paths import PF_CAP_NODES_REL, PF_REG_CATALOG_REL, resolve_pf_catalog_paths

PF_DATA_ROOT = None
if PHYSICS_WEIGHT > 0:
    _reg_cat, _cap_map, PF_DATA_ROOT = resolve_pf_catalog_paths(
        repo=REPO,
        preferred_root=None,
        chunk_parent=chunk_parent,
    )

print("=== Preflight (add-ons full training) ===")
print(f"REPO:           {REPO}")
print(f"SMOKE_TEST:     {SMOKE_TEST}")
print(f"SMOKE_CHUNK_COUNT: {SMOKE_CHUNK_COUNT}")
print(f"CHUNK_GLOB:     {CHUNK_GLOB}")
print(f"EPOCHS:         {EPOCHS}")
print(f"PATIENCE:       {PATIENCE}")
print(f"SEED:           {SEED}")
print(f"CHUNK_PARENT:   {chunk_parent}")
if not chunk_parent.is_dir():
    raise FileNotFoundError(f"CHUNK_PARENT not found: {chunk_parent}")

run_preview = sorted(
    p.name for p in chunk_parent.iterdir() if p.is_dir() and p.name.startswith("run_")
)[:5]
print(f"run_* preview:  {run_preview}")

if NODE_PE_CSV is None:
    node_pe = _find_node_pe_csv(chunk_parent)
else:
    node_pe = _resolve_data_path(NODE_PE_CSV, label="NODE_PE_CSV", colab_fallback=COLAB_NODE_PE_DEFAULT if _on_colab() else None)
    if not node_pe.is_file():
        warnings.warn(
            f"NODE_PE_CSV not found at {node_pe}; auto-discovering under CHUNK_PARENT.",
            UserWarning,
            stacklevel=2,
        )
        node_pe = _find_node_pe_csv(chunk_parent)

print(f"NODE_PE_CSV:    {node_pe}")
print(f"DA_CACHE_ROOT:  {da_cache_root}")
print(f"GNN_CACHE_ROOT: {gnn_cache_root}")
print(f"RUNS_PARENT:    {runs_parent}")

if PHYSICS_WEIGHT > 0:
    assert PF_DATA_ROOT is not None
    _pf_checks = [
        ("reg_catalog", PF_DATA_ROOT / PF_REG_CATALOG_REL),
        ("cap_nodes", PF_DATA_ROOT / PF_CAP_NODES_REL),
        ("electrical_distance", PF_DATA_ROOT / "electrical_distance_from_substation.csv"),
        (
            "hetero_mv_nodes",
            PF_DATA_ROOT / "Heterogenous GNN dataset/nodes/hetero_mv_nodes_load_transformer.csv",
        ),
        ("bus_kv_cache", PF_DATA_ROOT / "bus_kv_base_by_node.csv"),
    ]
    print(f"PF_DATA_ROOT:   {PF_DATA_ROOT}")
    for label, p in _pf_checks:
        status = "OK" if p.is_file() else "MISSING"
        print(f"  PF {label}: {status}  {p}")
        if not p.is_file():
            raise FileNotFoundError(f"Physics preflight missing {label}: {p}")
else:
    print("PF_DATA_ROOT:   (skipped — PHYSICS_WEIGHT=0)")

print("=================")

da_cache_root.mkdir(parents=True, exist_ok=True)
gnn_cache_root.mkdir(parents=True, exist_ok=True)
runs_parent.mkdir(parents=True, exist_ok=True)

chunks = _chunks_from_subdir_glob(chunk_parent, CHUNK_GLOB)
if not chunks:
    raise RuntimeError(f"No folders for CHUNK_GLOB={CHUNK_GLOB!r} under {chunk_parent}")

for p in chunks:
    for name in (
        "gnn_node_features_and_targets_mvagg.csv",
        "gnn_edges_phase_static.csv",
        "gnn_sample_meta.csv",
    ):
        if not (p / name).is_file():
            raise FileNotFoundError(f"Missing {name} in {p.name}")

print(f"Found {len(chunks)} chunk(s) for CHUNK_GLOB under {chunk_parent}")

tag = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
_pf_suffix = "_pf" if PHYSICS_WEIGHT > 0 else ""
_addon_run_suffix = "_addons_smoke" if SMOKE_TEST else "_addons_full"
out_dir = runs_parent / (
    f"da_gps_chunked_l{LIGHTER_LAYERS}_h{LIGHTER_HIDDEN}_mvagg_gine_metaaux_regce{_pf_suffix}{_addon_run_suffix}_{tag}"
)
out_dir.mkdir(parents=True, exist_ok=True)

_early_stop = "voltage" if PHYSICS_WEIGHT > 0 else "total"

cmd = [
    sys.executable, "-u", "train_da_gps_multitask_complex_voltage_gine.py",
    "--chunk_parent", str(chunk_parent),
    "--chunk_subdir_glob", CHUNK_GLOB,
    "--nodes_csv", "gnn_node_features_and_targets_mvagg.csv",
    "--edge_catalog_csv", "gnn_edges_phase_static.csv",
    "--meta_csv", "gnn_sample_meta.csv",
    "--node_feature_cols", "p_load_kw,q_load_kvar,p_pv_kw",
    "--exclude_bess_features",
    "--node_pe_csv", str(node_pe),
    "--node_pe_cols", "auto",
    "--n_system_tokens", "10",
    "--aux_meta_cols", META_AUX_COLS,
    "--lambda_pv", "0.1",
    "--out_dir", str(out_dir),
    "--cache_dir", str(da_cache_root),
    "--bootstrap_gnn_cache_dir", str(gnn_cache_root),
    "--epochs", str(EPOCHS),
    "--batch_size", "64",
    "--hidden", str(LIGHTER_HIDDEN),
    "--layers", str(LIGHTER_LAYERS),
    "--heads", str(LIGHTER_HEADS),
    "--node_emb_dim", str(LIGHTER_NODE_EMB_DIM),
    "--edge_emb_dim", str(LIGHTER_EDGE_EMB_DIM),
    "--lr", "5e-4",
    "--weight_decay", "1e-5",
    "--lambda_cap", "0.1",
    "--lambda_reg", "0.1",
    "--reg_loss", "ce",
    "--per_device_cap_head",
    "--per_device_reg_head",
    "--patience", str(PATIENCE),
    "--seed", str(SEED),
    "--train_frac", "0.80",
    "--val_frac", "0.10",
    "--sample_frac", "1.0",
    "--num_workers", str(NUM_WORKERS),
    "--log_every", "10",
    "--checkpoint_every", "10",
    "--early_stop_on", _early_stop,
    "--dropout", "0.1",
    "--attn_reg_territory_bias",
    "--attn_reg_territory_beta", "2.0",
    "--reg_ordinal_ce",
    "--volt_violation_weight",
    "--volt_lo_pu", "0.96",
    "--volt_hi_pu", "1.04",
    "--volt_violation_alpha", "2.0",
]

if PHYSICS_WEIGHT > 0:
    _pf_flags = [
        "--pf_data_root", str(PF_DATA_ROOT),
        "--loss_power_balance_weight", str(PHYSICS_WEIGHT),
        "--pf_sparse_y", "1",
        "--pf_huber_delta_kw", "10",
        "--pf_detach_controls",
    ]
    _bal = Path(PF_BALANCE_NODE_LIST_CSV)
    if not _bal.is_absolute():
        _bal = (REPO / _bal).resolve()
    if not _bal.is_file():
        raise FileNotFoundError(f"PF_BALANCE_NODE_LIST_CSV not found: {_bal}")
    _pf_flags.extend(["--pf_balance_node_list_csv", str(_bal)])
    cmd.extend(_pf_flags)

_mode = "physics-informed" if PHYSICS_WEIGHT > 0 else "baseline"
print(f"PHYSICS_WEIGHT={PHYSICS_WEIGHT} ({_mode})")
print(f"early_stop_on={_early_stop}")
print("\nRunning:\n ", " ".join(cmd), "\n", flush=True)

with subprocess.Popen(
    cmd,
    cwd=str(REPO),
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    bufsize=1,
    text=True,
) as proc:
    assert proc.stdout is not None
    for line in proc.stdout:
        print(line, end="")
    rc = proc.wait()

if rc != 0:
    raise subprocess.CalledProcessError(rc, cmd)

print("\nTraining completed.")
print("Run dir:", out_dir.resolve())
print("Checkpoint (best):", (out_dir / "da_gps_multitask_best.pt").resolve())
print("Checkpoint (last):", (out_dir / "training_last.pt").resolve())
print("Regulator classes:", (out_dir / "reg_class_tables.json").resolve())
print("Report:", (out_dir / "da_gps_report.json").resolve())
_run_name = out_dir.name
print("\n=== Section 8 prerequisite ===")
print(f"Baseline run folder:  {out_dir.resolve()}")
print(f"Section 8 expects:    {MYDRIVE_DATA / 'runs' / _run_name}")
print("Checkpoint for init:  da_gps_multitask_best.pt (or training_last.pt)")
if _on_colab() and not str(out_dir.resolve()).startswith(str(MYDRIVE_DATA.resolve())):
    print(
        "WARNING: run dir is NOT under MyDrive/datasets_gnn2 — "
        "copy to Drive before disconnecting the VM or Section 8 will fail."
    )
else:
    print("Checkpoints are on Drive; safe to run Section 8 after this session ends.")
_write_smoke_compare_manifest(
    runs_parent,
    out_dir=out_dir,
    physics_weight=PHYSICS_WEIGHT,
    chunk_glob=CHUNK_GLOB,
    n_chunks=len(chunks),
    epochs=EPOCHS,
    seed=SEED,
)
````
